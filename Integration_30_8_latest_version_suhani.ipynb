```python
#CELL 1
# ---------------- Install Required Packages ----------------
!pip install -U bitsandbytes transformers diffusers accelerate safetensors

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt
import gc

# ---------------- Model Name ----------------
model_name = 'filipealmeida/Mistral-7B-Instruct-v0.1-sharded'

# ---------------- Load Quantized Model ----------------
def load_quantized_model(model_name):
    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_use_double_quant=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_compute_dtype=torch.bfloat16,
        llm_int8_enable_fp32_cpu_offload=True
    )
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        quantization_config=bnb_config,
        device_map="auto",
        trust_remote_code=True
    )
    return model

# ---------------- Initialize Tokenizer ----------------
def initialize_tokenizer(model_name: str):
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    if tokenizer.bos_token_id is None:
        tokenizer.bos_token_id = tokenizer.cls_token_id or 1
    if tokenizer.eos_token_id is None:
        tokenizer.eos_token_id = tokenizer.sep_token_id or 2
    # Fix: Add padding token
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
    return tokenizer

# ---------------- Build Chat Prompt ----------------
def build_chat_prompt(user_text: str):
    return (
        "You are a meticulous ad-background generator. "
        "Follow ALL constraints exactly. "
        "Output only the 5-line background description.\n\n"
        f"{user_text}"
    )

# ---------------- Generate Response ----------------
def generate_response(prompt_text: str, model, tokenizer):
    device = "cuda" if torch.cuda.is_available() else "cpu"
    chat_prompt = build_chat_prompt(prompt_text)

    inputs = tokenizer(chat_prompt, return_tensors="pt", padding=True, truncation=True, max_length=512).to(device)

    torch.manual_seed(0)  # reproducibility
    with torch.no_grad():  # Fix: Add no_grad for memory efficiency
        output = model.generate(
            **inputs,
            do_sample=True,  # Fix: Enable sampling for better creativity
            temperature=0.7,  # Fix: Add some temperature for variation
            max_new_tokens=200,
            eos_token_id=tokenizer.eos_token_id,
            pad_token_id=tokenizer.pad_token_id  # Fix: Add pad_token_id
        )

    # Fix: Only decode the generated part
    generated_tokens = output[0][inputs['input_ids'].shape[1]:]
    text = tokenizer.decode(generated_tokens, skip_special_tokens=True)
    return text

# ---------------- Enforce Exactly 5 Lines ----------------
def enforce_5_lines(text: str):
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]

    # If we get one long paragraph, try to split it into sentences
    if len(lines) == 1 and len(lines[0]) > 100:
        # Split by periods and clean up
        sentences = [s.strip() for s in lines[0].split('.') if s.strip()]
        lines = sentences

    # If we still don't have enough lines, split long sentences further
    if len(lines) < 5:
        new_lines = []
        for line in lines:
            if len(line) > 80 and ',' in line:
                # Split by commas for long sentences
                parts = [part.strip() for part in line.split(',') if part.strip()]
                new_lines.extend(parts[:2])  # Take first 2 parts
            else:
                new_lines.append(line)
        lines = new_lines

    # If still not enough, create meaningful background elements based on the content
    if len(lines) < 5:
        base_elements = [
            "Soft gradient transitions create depth and dimension",
            "Subtle lighting enhances the overall composition",
            "Abstract patterns add visual interest without distraction",
            "Professional photography aesthetic maintains focus",
            "Clean minimal design complements the product presentation"
        ]

        # Add meaningful elements to reach 5 lines
        while len(lines) < 5:
            lines.append(base_elements[len(lines) - 1] if len(lines) - 1 < len(base_elements) else base_elements[-1])

    return lines[:5]

# ---------------- Build Product Prompt - Made Dynamic ----------------
def generate_prompt_for_ad_background(data, color, motif):
    # Fix: Make it completely dynamic based on user data
    description_parts = []

    # Dynamically build description from any data provided
    for key, value in data.items():
        if value and str(value).strip():  # Only include non-empty values
            if key.lower() in ['gender']:
                description_parts.append(f"designed for {str(value).lower()}")
            elif key.lower() in ['basecolour', 'color', 'colour']:
                description_parts.append(f"features a {str(value).lower()} color scheme")
            elif key.lower() in ['usage', 'occasion']:
                description_parts.append(f"intended for {str(value).lower()} use")
            elif key.lower() in ['mastercategory', 'category']:
                description_parts.append(f"part of the {str(value).lower()} collection")
            elif key.lower() in ['subcategory', 'type']:
                description_parts.append(f"specifically a {str(value).lower()} item")
            elif key.lower() in ['year', 'season']:
                description_parts.append(f"from {str(value)}")
            else:
                # For any other attributes, include them generically
                description_parts.append(f"has {key.lower()} of {str(value).lower()}")

    # Build product description
    if description_parts:
        product_description = "This product " + ", ".join(description_parts) + "."
    else:
        product_description = "This is a product that needs an advertising background."

    return f"""
Context:
{product_description}

Constraints (MUST follow exactly):
- Background color scheme should be based on {color} tones.
- Design motif should be inspired by {motif} but ONLY as abstract design elements, patterns, or shapes.
- NO realistic objects, people, furniture, or interior/room elements of any kind.
- NO literal representations of {motif} - only abstract interpretations.
- Must be a clean, professional advertising background that complements the product without competing for attention.
- Abstract, minimalist fashion-ad mood only (gradients, soft lighting, reflective surfaces, subtle geometric patterns).
- Background should enhance the product presentation, not distract from it.
- Output EXACTLY 5 separate lines, each on its own line.

Output format (CRITICAL):
- Line 1: [describe the main background color/gradient]
- Line 2: [describe the abstract design elements or patterns]
- Line 3: [describe the lighting and atmosphere]
- Line 4: [describe texture or surface qualities]
- Line 5: [describe overall mood and professional appeal]
- Each line must be separate and concise (under 15 words)
- NO paragraphs, NO combining lines, NO additional text

Begin now:
""".strip()

# ---------------- Convert 5-line text to SD-friendly prompt ----------------
def text_to_visual_prompt(lines, color, motif):
    """
    Converts the 5-line textual background description into a concise
    prompt under 77 tokens for Stable Diffusion - focused on abstract backgrounds.
    """
    # Base elements for professional ad background
    base_elements = [
        f"{color} color scheme",
        "abstract background",
        "professional lighting",
        "minimalist design",
        "commercial photography style",
        "clean aesthetic"
    ]

    # Convert motif to abstract design element
    if "flower" in motif.lower():
        base_elements.append("subtle organic patterns")
    elif "geometric" in motif.lower():
        base_elements.append("geometric patterns")
    elif "wave" in motif.lower():
        base_elements.append("flowing curves")
    else:
        # Generic abstract interpretation
        base_elements.append("abstract design elements")

    # Extract abstract visual elements from generated lines
    abstract_keywords = ['gradient', 'glow', 'shine', 'smooth', 'flowing', 'soft', 'subtle', 'elegant', 'refined']

    for line in lines:
        line_lower = line.lower()
        for keyword in abstract_keywords:
            if keyword in line_lower and keyword not in base_elements:
                base_elements.append(keyword)

    # Keep it focused and under token limit
    visual_prompt = ", ".join(base_elements[:8])
    return visual_prompt

# ---------------- Memory Cleanup Function ----------------
def cleanup_memory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

# ---------------- Main Generation Function ----------------
def generate_ad_background_for_product(product_data, background_color, motif):
    """
    Main function that takes any product data and generates background
    """
    print("=== Starting Ad Background Generation ===")
    print(f"Product Data: {product_data}")
    print(f"Background Color: {background_color}")
    print(f"Motif: {motif}")

    # Load models
    print("Loading text generation model...")
    model = load_quantized_model(model_name)
    tokenizer = initialize_tokenizer(model_name)

    # Generate prompt
    print("Generating background description...")
    ad_prompt = generate_prompt_for_ad_background(product_data, background_color, motif)

    # Generate response
    raw_output = generate_response(ad_prompt, model, tokenizer)
    ad_lines = enforce_5_lines(raw_output)

    print("Generated Ad Background Description:")
    for i, line in enumerate(ad_lines, 1):
        print(f"{i}. {line}")

    # Create negative prompt dynamically - Enhanced to prevent literal objects
    excluded_terms = []
    for key, value in product_data.items():
        if key.lower() in ['subcategory', 'mastercategory', 'category', 'type'] and value:
            excluded_terms.append(str(value).lower())

    # Enhanced negative prompt to prevent literal interpretations
    negative_prompt = "text, letters, words, logo, watermark, signature, human, person, object, product, realistic, photorealistic, literal, flowers, plants, animals, furniture, interior, room, buildings"
    if excluded_terms:
        negative_prompt += ", " + ", ".join(excluded_terms)

    # Add motif-specific negatives to prevent literal interpretation
    if "flower" in motif.lower():
        negative_prompt += ", realistic flowers, petals, stems, leaves, garden, nature"
    elif "wave" in motif.lower():
        negative_prompt += ", ocean, water, sea, realistic waves"
    elif "star" in motif.lower():
        negative_prompt += ", realistic stars, night sky, space"

    # Convert to SD prompt
    sd_prompt = text_to_visual_prompt(ad_lines, background_color, motif)
    print(f"\nStable Diffusion Prompt: {sd_prompt}")
    print(f"Negative Prompt: {negative_prompt}")

    # Free text model memory
    print("Freeing text model memory...")
    model.cpu()
    del model
    cleanup_memory()

    # Load Stable Diffusion
    print("Loading Stable Diffusion pipeline...")
    pipe = StableDiffusionPipeline.from_pretrained(
        "runwayml/stable-diffusion-v1-5",
        torch_dtype=torch.float16
    ).to("cuda")

    # Generate image
    print("Generating background image...")
    pipe.safety_checker = None  # optional: disable safety check if safe

    background_image = pipe(
        prompt=sd_prompt,
        negative_prompt=negative_prompt,
        guidance_scale=8.0,
        num_inference_steps=50
    ).images[0]

    # Save and display
    background_image.save("generated_background.png")
    print("Background generated successfully!")

    plt.figure(figsize=(10, 8))
    plt.imshow(background_image)
    plt.axis("off")
    plt.title("Generated Background")
    plt.show()

    return background_image, ad_lines, sd_prompt

# ---------------- Example Usage ----------------
if __name__ == "__main__":
    # Example 1: Your original watch data
    product_data = {
        "gender": "women",
        "baseColour": "silver",
        "usage": "casual",
        "subCategory": "watches",
        "masterCategory": "accessories",
        "year": 2016
    }

    generate_ad_background_for_product(
        product_data=product_data,
        background_color="pink",
        motif="blue flowers"
    )

    # Example 2: Different product to show flexibility
    print("\n" + "="*60)
    print("Example 2: Different product")

    product_data_2 = {
        "gender": "men",
        "color": "black",
        "occasion": "formal",
        "type": "shoes",
        "brand": "luxury",
        "material": "leather"
    }

    generate_ad_background_for_product(
        product_data=product_data_2,
        background_color="deep navy",
        motif="geometric patterns"
    )
```

```python

import matplotlib.pyplot as plt
import json


from PIL import Image, ImageDraw, ImageFilter

def create_cylindrical_pedestal(width, height, pedestal_color="white", shadow_intensity=0.3):
    """Creates a more realistic cylindrical pedestal with gradient and soft shadow."""
    img = Image.new('RGBA', (width, height), (0, 0, 0, 0))
    draw = ImageDraw.Draw(img)

    pedestal_width = int(width * 0.35)
    pedestal_height = int(height * 0.22)

    center_x = width // 2
    center_y = int(height * 0.72)

    # Top ellipse
    top_ellipse_coords = [
        center_x - pedestal_width // 2,
        center_y - pedestal_height // 2 - 12,
        center_x + pedestal_width // 2,
        center_y - pedestal_height // 2 + 12
    ]

    # Gradient fill (vertical)
    body = Image.new("RGBA", (pedestal_width, pedestal_height), (0, 0, 0, 0))
    body_draw = ImageDraw.Draw(body)
    for y in range(pedestal_height):
        shade = int(255 - (y / pedestal_height) * 40)  # subtle darker bottom
        body_draw.line([(0, y), (pedestal_width, y)], fill=(shade, shade, shade, 255))
    img.paste(body, (center_x - pedestal_width // 2, center_y - pedestal_height // 2), body)

    # Top ellipse with subtle highlight
    draw.ellipse(top_ellipse_coords, fill=(250, 250, 250, 255), outline=(200, 200, 200, 255), width=2)

    # Soft shadow under pedestal
    shadow = Image.new("RGBA", img.size, (0, 0, 0, 0))
    shadow_draw = ImageDraw.Draw(shadow)
    shadow_ellipse = [
        center_x - pedestal_width // 2,
        center_y + pedestal_height // 2 - 5,
        center_x + pedestal_width // 2,
        center_y + pedestal_height // 2 + 10
    ]
    shadow_draw.ellipse(shadow_ellipse, fill=(0, 0, 0, int(255 * shadow_intensity)))
    shadow = shadow.filter(ImageFilter.GaussianBlur(6))
    img = Image.alpha_composite(shadow, img)

    # Metadata stays the same
    img.platform_info = {
        'center': (center_x, center_y - pedestal_height // 2),
        'top_surface': {
            'bbox': top_ellipse_coords,
            'width': pedestal_width,
            'height': 24
        },
        'type': 'cylindrical_pedestal'
    }

    return img


def create_cube_pedestal(width, height, pedestal_color="white", perspective_angle=15):
    """Creates a cube pedestal with metadata for placement."""
    img = Image.new('RGBA', (width, height), (0, 0, 0, 0))
    draw = ImageDraw.Draw(img)

    cube_size = int(min(width, height) * 0.18)
    center_x = width // 2
    center_y = int(height * 0.75)
    offset = int(cube_size * 0.3)

    top_color = (255, 255, 255, 255)
    front_color = (255, 255, 255, 255)
    side_color = (245, 245, 245, 255)
    outline_color = (220, 220, 220, 255)

    # Shadow for realism
    shadow_points = [
        (center_x - cube_size // 2 + 8, center_y + cube_size // 2 + 8),
        (center_x + cube_size // 2 + 8, center_y + cube_size // 2 + 8),
        (center_x + cube_size // 2 + offset + 8, center_y + cube_size // 2 - offset + 8),
        (center_x - cube_size // 2 + offset + 8, center_y + cube_size // 2 - offset + 8)
    ]
    draw.polygon(shadow_points, fill=(0, 0, 0, 80))

    # Cube faces
    front_points = [
        (center_x - cube_size // 2, center_y - cube_size // 2),
        (center_x + cube_size // 2, center_y - cube_size // 2),
        (center_x + cube_size // 2, center_y + cube_size // 2),
        (center_x - cube_size // 2, center_y + cube_size // 2)
    ]
    side_points = [
        (center_x + cube_size // 2, center_y - cube_size // 2),
        (center_x + cube_size // 2 + offset, center_y - cube_size // 2 - offset),
        (center_x + cube_size // 2 + offset, center_y + cube_size // 2 - offset),
        (center_x + cube_size // 2, center_y + cube_size // 2)
    ]
    top_points = [
        (center_x - cube_size // 2, center_y - cube_size // 2),
        (center_x + cube_size // 2, center_y - cube_size // 2),
        (center_x + cube_size // 2 + offset, center_y - cube_size // 2 - offset),
        (center_x - cube_size // 2 + offset, center_y - cube_size // 2 - offset)
    ]

    # Draw cube
    draw.polygon(side_points, fill=side_color, outline=outline_color)
    draw.polygon(top_points, fill=top_color, outline=outline_color)
    draw.polygon(front_points, fill=front_color, outline=outline_color)

    # Metadata for placement
    img.platform_info = {
        'center': (center_x, center_y - cube_size // 2),
        'top_surface': {
            'points': top_points,
            'width': cube_size + offset,
            'height': cube_size // 2
        },
        'type': 'cube_pedestal'
    }

    return img


def add_pedestal_to_background(background_image, pedestal_type="cylinder", pedestal_color="white", pedestal_style="modern"):
    bg_img = background_image.copy() if hasattr(background_image, 'size') else Image.fromarray(background_image)
    width, height = bg_img.size

    pedestal_img = create_cylindrical_pedestal(width, height, pedestal_color) if pedestal_type == "cylinder" \
        else create_cube_pedestal(width, height, pedestal_color)

    final_img = Image.alpha_composite(bg_img.convert('RGBA'), pedestal_img).convert('RGB')

    if hasattr(pedestal_img, 'platform_info'):
        final_img.platform_info = pedestal_img.platform_info

    return final_img


def get_platform_placement_info(image_with_pedestal):
    """Returns clean metadata for consistent placement."""
    if hasattr(image_with_pedestal, 'platform_info'):
        platform_info = image_with_pedestal.platform_info
        return {
            'center': platform_info['center'],
            'top_surface': platform_info['top_surface'],
            'type': f"generated_{platform_info['type']}"
        }
    return None


def generate_ad_background_with_pedestal(product_data, background_color, motif,
                                         pedestal_type="cylinder", pedestal_color="white",
                                         pedestal_style="modern", show_guide=True):
    print("=== Starting Ad Background Generation with Pure White Pedestal ===")

    # Use background generator from Cell 1
    background_image, ad_lines, sd_prompt = generate_ad_background_for_product(
        product_data, background_color, motif
    )

    # Add pedestal
    print(f"Adding solid white {pedestal_type} pedestal...")
    background_with_pedestal = add_pedestal_to_background(
        background_image, pedestal_type, pedestal_color, pedestal_style
    )

    background_with_pedestal.save("background_with_solid_pedestal.png")

    # Save metadata
    platform_info = get_platform_placement_info(background_with_pedestal)
    if platform_info:
        with open("platform_metadata.json", "w") as f:
            json.dump(platform_info, f, indent=2)
        print("Platform metadata saved to platform_metadata.json")

    print("Pure white pedestal generated successfully!")

    if show_guide:
        fig, axes = plt.subplots(1, 2, figsize=(12, 5))
        axes[0].imshow(background_image)
        axes[0].set_title("Original Background")
        axes[0].axis("off")

        axes[1].imshow(background_with_pedestal)
        axes[1].set_title("With Pure White Pedestal")
        axes[1].axis("off")

        plt.tight_layout()
        plt.show()

    return background_with_pedestal, ad_lines, sd_prompt


# ---------------- Usage Examples ----------------
def example_usage():
    """
    Examples of how to use the enhanced function
    """
    # Your original product data
    product_data = {
        "gender": "women",
        "baseColour": "silver",
        "usage": "casual",
        "subCategory": "watches",
        "masterCategory": "accessories",
        "year": 2016
    }

    # Example 1: Cylinder pedestal (like your reference image)
    print("Example 1: Cylinder Pedestal")
    generate_ad_background_with_pedestal(
        product_data=product_data,
        background_color="turquoise",  # To match your reference image
        motif="geometric circles",
        pedestal_type="cylinder",
        pedestal_color="white",
        pedestal_style="modern"
    )

    # # Example 2: Geometric pedestal
    # print("\n" + "="*50)
    # print("Example 2: Geometric Pedestal")
    # generate_ad_background_with_pedestal(
    #     product_data=product_data,
    #     background_color="pink",
    #     motif="blue flowers",
    #     pedestal_type="geometric",
    #     pedestal_color="glass",
    #     pedestal_style="modern"
    # )

# To integrate with your existing code, simply replace your final function call with:
# generate_ad_background_with_pedestal(product_data, background_color, motif)

if __name__ == "__main__":
    example_usage()
```

```python
#CELL 1
# ---------------- Install Required Packages ----------------
!pip install -U bitsandbytes transformers diffusers accelerate safetensors

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt
import gc

# ---------------- Model Name ----------------
model_name = 'filipealmeida/Mistral-7B-Instruct-v0.1-sharded'

# ---------------- Load Quantized Model ----------------
def load_quantized_model(model_name):
    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_use_double_quant=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_compute_dtype=torch.bfloat16,
        llm_int8_enable_fp32_cpu_offload=True
    )
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        quantization_config=bnb_config,
        device_map="auto",
        trust_remote_code=True
    )
    return model

# ---------------- Initialize Tokenizer ----------------
def initialize_tokenizer(model_name: str):
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    if tokenizer.bos_token_id is None:
        tokenizer.bos_token_id = tokenizer.cls_token_id or 1
    if tokenizer.eos_token_id is None:
        tokenizer.eos_token_id = tokenizer.sep_token_id or 2
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
    return tokenizer

# ---------------- Enhanced Product Analysis ----------------
def analyze_product_for_appeal(data):
    """Analyze product data to determine attractive design elements"""
    appeal_factors = {
        'luxury_level': 'standard',
        'target_emotion': 'confident',
        'visual_style': 'modern',
        'energy_level': 'medium'
    }

    # Analyze luxury level
    if any(term in str(data).lower() for term in ['luxury', 'premium', 'gold', 'diamond', 'designer']):
        appeal_factors['luxury_level'] = 'luxury'
    elif any(term in str(data).lower() for term in ['casual', 'everyday', 'basic']):
        appeal_factors['luxury_level'] = 'accessible'

    # Analyze target emotion based on usage/occasion
    usage_str = str(data.get('usage', '') + ' ' + data.get('occasion', '')).lower()
    if any(term in usage_str for term in ['formal', 'business', 'professional']):
        appeal_factors['target_emotion'] = 'sophisticated'
    elif any(term in usage_str for term in ['party', 'celebration', 'festive']):
        appeal_factors['target_emotion'] = 'exciting'
    elif any(term in usage_str for term in ['sport', 'fitness', 'active']):
        appeal_factors['target_emotion'] = 'energetic'
    elif any(term in usage_str for term in ['casual', 'daily', 'everyday']):
        appeal_factors['target_emotion'] = 'comfortable'

    # Analyze visual style based on category and year
    category_str = str(data.get('masterCategory', '') + ' ' + data.get('subCategory', '')).lower()
    year = data.get('year', 2020)

    if 'accessories' in category_str:
        appeal_factors['visual_style'] = 'elegant'
    elif 'apparel' in category_str:
        appeal_factors['visual_style'] = 'fashionable'
    elif 'footwear' in category_str:
        appeal_factors['visual_style'] = 'dynamic'

    # Adjust for recency
    if isinstance(year, (int, str)) and int(year) >= 2020:
        appeal_factors['energy_level'] = 'high'

    return appeal_factors

# ---------------- Enhanced Attractive Background Prompt Builder ----------------
def build_attractive_background_prompt(data, color, motif):
    """Build prompt focused on creating attractive, customer-drawing backgrounds"""

    appeal_factors = analyze_product_for_appeal(data)

    # Create dynamic product context
    product_context = []
    for key, value in data.items():
        if value and str(value).strip():
            product_context.append(f"{key}: {value}")

    context_str = ", ".join(product_context) if product_context else "fashion product"

    # Enhanced attractive elements based on appeal factors
    attractive_elements = {
        'luxury': [
            "premium metallic accents and sophisticated highlights",
            "rich depth with layered textures and premium finishes",
            "elegant bokeh effects with soft luminous particles",
            "sophisticated color transitions with metallic undertones",
            "high-end commercial photography aesthetic with luxury appeal"
        ],
        'accessible': [
            "vibrant color pops with energetic gradient flows",
            "playful abstract shapes and dynamic visual movement",
            "bright atmospheric lighting with cheerful highlights",
            "smooth flowing textures with contemporary appeal",
            "modern lifestyle photography mood with approachable warmth"
        ],
        'standard': [
            "balanced color harmonies with attractive depth",
            "stylish abstract elements and modern design patterns",
            "professional studio lighting with appealing highlights",
            "refined surface textures with contemporary finish",
            "commercial photography aesthetic with broad appeal"
        ]
    }

    # Emotion-based visual enhancements
    emotion_styles = {
        'sophisticated': "elegant refinement",
        'exciting': "dynamic energy",
        'energetic': "vibrant movement",
        'comfortable': "warm inviting",
        'confident': "bold striking"
    }

    style_modifier = emotion_styles.get(appeal_factors['target_emotion'], 'attractive modern')
    base_template = attractive_elements.get(appeal_factors['luxury_level'], attractive_elements['standard'])

    return f"""
You are an expert advertising background designer creating visually compelling backgrounds that attract customers and drive sales.

Product Context: {context_str}
Target Appeal: {appeal_factors['target_emotion']} and {appeal_factors['luxury_level']} positioning

Create an ATTRACTIVE advertising background with these specifications:
- Primary color scheme: {color} with complementary accent tones
- Design inspiration: Abstract interpretation of {motif} as visual design elements
- Style mood: {style_modifier} commercial photography aesthetic
- Must be eye-catching and customer-appealing while maintaining product focus

CRITICAL REQUIREMENTS:
- Create 5 distinct visual layers that build an attractive composition
- Include specific attractive elements: lighting effects, color dynamics, visual interest
- Make it appealing enough to stop customers and draw attention
- Professional but NOT boring - needs visual magnetism
- Abstract design elements only (no literal objects)

Output exactly 5 lines describing the attractive background:
Line 1: Main color scheme and attractive gradient system
Line 2: Eye-catching design elements inspired by {motif} (abstract only)
Line 3: Dynamic lighting and atmospheric effects that draw attention
Line 4: Appealing textures and surface qualities that add visual interest
Line 5: Overall mood that creates customer attraction and purchase appeal

Each line under 20 words, focus on ATTRACTIVE and APPEALING elements:
"""

# ---------------- Enhanced Visual Prompt Converter ----------------
def create_attractive_sd_prompt(lines, color, motif, appeal_factors):
    """Convert 5-line description to SD prompt optimized for attractive results"""

    # Base attractive elements
    attractive_base = [
        "professional advertising background",
        f"attractive {color} color palette",
        "high-end commercial photography",
        "studio lighting setup"
    ]

    # Add appeal-specific elements
    if appeal_factors['luxury_level'] == 'luxury':
        attractive_base.extend([
            "premium metallic accents",
            "sophisticated bokeh effects",
            "luxury commercial aesthetic"
        ])
    elif appeal_factors['luxury_level'] == 'accessible':
        attractive_base.extend([
            "vibrant color dynamics",
            "energetic visual flow",
            "contemporary lifestyle aesthetic"
        ])
    else:
        attractive_base.extend([
            "balanced visual appeal",
            "modern design elements",
            "commercial photography style"
        ])

    # Convert motif to attractive abstract elements
    motif_elements = {
        'flower': 'organic flowing patterns, petal-inspired curves',
        'geometric': 'dynamic geometric patterns, angular design elements',
        'wave': 'flowing wave patterns, fluid design elements',
        'star': 'radial burst patterns, stellar design elements',
        'circle': 'circular design patterns, ring elements',
        'nature': 'organic abstract patterns, natural flow elements'
    }

    # Find best motif match
    motif_addition = "abstract design patterns"
    for key, value in motif_elements.items():
        if key in motif.lower():
            motif_addition = value
            break

    attractive_base.append(motif_addition)

    # Extract appealing keywords from generated lines
    appeal_keywords = []
    for line in lines:
        line_lower = line.lower()
        appealing_terms = [
            'shimmer', 'glow', 'radiant', 'vibrant', 'dynamic', 'flowing',
            'elegant', 'sophisticated', 'premium', 'luxurious', 'striking',
            'captivating', 'alluring', 'magnetic', 'brilliant', 'stunning'
        ]
        for term in appealing_terms:
            if term in line_lower:
                appeal_keywords.append(term)

    # Combine all elements (keep under 77 tokens)
    final_elements = attractive_base + appeal_keywords[:3]
    sd_prompt = ", ".join(final_elements[:12])  # Limit to prevent token overflow

    return sd_prompt

# ---------------- Enhanced Negative Prompt ----------------
def create_smart_negative_prompt(product_data, motif):
    """Create intelligent negative prompt that prevents boring results while avoiding literal objects"""

    # Base negatives - prevent literal interpretations but allow attractive abstractions
    base_negatives = [
        "text", "letters", "words", "logo", "watermark", "signature",
        "human", "person", "face", "body", "hands",
        "realistic objects", "product placement", "literal representation",
        "boring", "plain", "empty", "bland", "colorless"
    ]

    # Motif-specific negatives (prevent literal but allow abstract)
    motif_negatives = {
        'flower': ["realistic flowers", "actual petals", "garden", "plants", "stems", "leaves"],
        'wave': ["ocean", "water", "sea", "realistic waves", "beach"],
        'star': ["night sky", "space", "realistic stars", "astronomy"],
        'geometric': ["mathematical symbols", "blueprints", "technical drawings"],
        'nature': ["landscape", "trees", "grass", "realistic nature"]
    }

    # Add motif-specific negatives
    for key, negatives in motif_negatives.items():
        if key in motif.lower():
            base_negatives.extend(negatives)

    # Product-specific negatives to prevent literal product appearance
    product_negatives = []
    for key, value in product_data.items():
        if key.lower() in ['subcategory', 'mastercategory'] and value:
            product_negatives.append(f"realistic {str(value).lower()}")

    all_negatives = base_negatives + product_negatives
    return ", ".join(all_negatives[:15])  # Limit length

# ---------------- Enforce Exactly 5 Attractive Lines ----------------
def enforce_5_attractive_lines(text: str, color, motif):
    """Ensure exactly 5 lines with attractive, appealing content"""
    lines = [ln.strip() for ln in text.splitlines() if ln.strip()]

    # If we get one long paragraph, split by sentences
    if len(lines) == 1 and len(lines[0]) > 100:
        sentences = [s.strip() + "." for s in lines[0].split('.') if s.strip()]
        lines = sentences

    # If still not enough, split by commas for very long sentences
    if len(lines) < 5:
        new_lines = []
        for line in lines:
            if len(line) > 80 and ',' in line:
                parts = [part.strip() for part in line.split(',')]
                new_lines.extend(parts[:2])
            else:
                new_lines.append(line)
        lines = new_lines

    # If still insufficient, generate attractive template lines
    if len(lines) < 5:
        attractive_templates = [
            f"Stunning {color} gradients cascade with sophisticated depth and visual magnetism",
            f"Dynamic {motif}-inspired patterns create captivating visual flow and movement",
            "Brilliant studio lighting produces radiant highlights and dramatic atmospheric appeal",
            "Luxurious surface textures shimmer with premium commercial photography finish",
            "Irresistible visual composition designed to captivate customers and drive engagement"
        ]

        # Fill remaining slots with attractive templates
        while len(lines) < 5:
            template_index = len(lines) if len(lines) < len(attractive_templates) else 4
            lines.append(attractive_templates[template_index])

    return lines[:5]

# ---------------- Enhanced Chat Prompt for Attractiveness ----------------
def build_attractive_chat_prompt(user_text: str):
    return (
        "You are an expert advertising background designer specializing in creating "
        "visually compelling, customer-attracting backgrounds for commercial photography. "
        "Your backgrounds must be ATTRACTIVE, EYE-CATCHING, and designed to draw customer attention "
        "while maintaining professional commercial appeal. "
        "Focus on creating visual magnetism that converts browsers into buyers.\n\n"
        f"{user_text}"
    )

# ---------------- Enhanced Response Generation ----------------
def generate_attractive_response(prompt_text: str, model, tokenizer):
    device = "cuda" if torch.cuda.is_available() else "cpu"
    chat_prompt = build_attractive_chat_prompt(prompt_text)

    inputs = tokenizer(chat_prompt, return_tensors="pt", padding=True, truncation=True, max_length=512).to(device)

    torch.manual_seed(42)  # Changed seed for variation
    with torch.no_grad():
        output = model.generate(
            **inputs,
            do_sample=True,
            temperature=0.4,  # Increased for more creativity
            top_p=0.9,        # Added nucleus sampling for better quality
            max_new_tokens=250,  # Increased token limit
            eos_token_id=tokenizer.eos_token_id,
            pad_token_id=tokenizer.pad_token_id
        )

    generated_tokens = output[0][inputs['input_ids'].shape[1]:]
    text = tokenizer.decode(generated_tokens, skip_special_tokens=True)
    return text

# ---------------- Enhanced Product Appeal Analysis ----------------
def get_customer_psychology_elements(data, color, motif):
    """Analyze what psychological elements will attract customers"""

    psychology_map = {
        'gender': {
            'women': ['elegant', 'sophisticated', 'glamorous', 'refined'],
            'men': ['bold', 'powerful', 'confident', 'striking'],
            'unisex': ['versatile', 'modern', 'stylish', 'appealing']
        },
        'usage': {
            'formal': ['prestigious', 'authoritative', 'impressive', 'commanding'],
            'casual': ['comfortable', 'approachable', 'relaxed', 'friendly'],
            'party': ['exciting', 'vibrant', 'celebratory', 'festive'],
            'sports': ['energetic', 'dynamic', 'motivating', 'powerful']
        },
        'category': {
            'accessories': ['aspirational', 'desirable', 'coveted', 'attractive'],
            'apparel': ['fashionable', 'trendy', 'stylish', 'appealing'],
            'footwear': ['confident', 'bold', 'statement-making', 'impressive']
        }
    }

    appeal_words = []

    # Extract psychological triggers from product data
    for key, value in data.items():
        if key.lower() in psychology_map and value:
            value_str = str(value).lower()
            for trigger_key, trigger_words in psychology_map[key.lower()].items():
                if trigger_key in value_str:
                    appeal_words.extend(trigger_words[:2])

    # Add color psychology
    color_psychology = {
        'blue': ['trustworthy', 'calming', 'professional'],
        'red': ['passionate', 'bold', 'energetic'],
        'green': ['fresh', 'natural', 'harmonious'],
        'pink': ['romantic', 'gentle', 'charming'],
        'purple': ['luxurious', 'creative', 'mysterious'],
        'gold': ['prestigious', 'valuable', 'exclusive'],
        'silver': ['modern', 'sleek', 'sophisticated'],
        'black': ['elegant', 'powerful', 'premium'],
        'white': ['pure', 'clean', 'minimalist']
    }

    for color_key, emotions in color_psychology.items():
        if color_key in color.lower():
            appeal_words.extend(emotions[:2])

    return list(set(appeal_words))  # Remove duplicates

# ---------------- Enhanced Prompt Generation ----------------
def generate_attractive_prompt_for_ad_background(data, color, motif):
    """Generate prompt optimized for customer attraction"""

    appeal_factors = analyze_product_for_appeal(data)
    psychology_elements = get_customer_psychology_elements(data, color, motif)

    # Build dynamic product description
    product_description = "This product "
    description_parts = []

    for key, value in data.items():
        if value and str(value).strip():
            if key.lower() == 'gender':
                description_parts.append(f"targets {str(value).lower()} customers")
            elif key.lower() in ['basecolour', 'color']:
                description_parts.append(f"features {str(value).lower()} as primary color")
            elif key.lower() in ['usage', 'occasion']:
                description_parts.append(f"designed for {str(value).lower()} occasions")
            elif key.lower() in ['mastercategory', 'subcategory']:
                description_parts.append(f"belongs to {str(value).lower()} category")

    if description_parts:
        product_description += ", ".join(description_parts) + "."
    else:
        product_description += "needs an attractive advertising background."

    return f"""
MISSION: Create a visually MAGNETIC advertising background that makes customers stop, look, and want to buy.

Product Analysis:
{product_description}
Target Psychology: {', '.join(psychology_elements[:4])}
Appeal Level: {appeal_factors['luxury_level']} with {appeal_factors['target_emotion']} emotional targeting

VISUAL REQUIREMENTS (Must be CUSTOMER-ATTRACTIVE):
- Color Foundation: {color} as primary with strategic accent colors for maximum appeal
- Design Elements: Abstract {motif} interpretation that creates visual interest and desire
- Customer Psychology: Must trigger {appeal_factors['target_emotion']} emotions and {appeal_factors['luxury_level']} aspirations
- Visual Impact: Eye-catching enough to stop social media scrolling and capture attention
- Commercial Appeal: Professional but with strong attractive magnetism

STRICT OUTPUT FORMAT - EXACTLY 5 LINES:
Each line must describe attractive visual elements that will draw customers:

Line 1: [Main color scheme with attractive depth and visual appeal]
Line 2: [Eye-catching abstract design elements inspired by {motif}]
Line 3: [Compelling lighting effects and atmospheric drama]
Line 4: [Appealing textures and finishes that suggest quality]
Line 5: [Overall customer-attracting mood and commercial magnetism]

REMEMBER: This background must be attractive enough to make customers WANT the product!
Generate the 5 lines now:
""".strip()

# ---------------- Enhanced Main Generation Function ----------------
def generate_attractive_ad_background(product_data, background_color, motif):
    """
    Enhanced function that creates genuinely attractive, customer-drawing backgrounds
    """
    print("=== Enhanced Attractive Background Generation ===")
    print(f"Product Data: {product_data}")
    print(f"Background Color: {background_color}")
    print(f"Motif: {motif}")

    # Analyze product appeal
    appeal_factors = analyze_product_for_appeal(product_data)
    psychology_elements = get_customer_psychology_elements(product_data, background_color, motif)

    print(f"Appeal Analysis: {appeal_factors}")
    print(f"Customer Psychology: {psychology_elements}")

    # Load models
    print("Loading text generation model...")
    model = load_quantized_model(model_name)
    tokenizer = initialize_tokenizer(model_name)

    # Generate attractive background description
    print("Generating attractive background description...")
    attractive_prompt = generate_attractive_prompt_for_ad_background(product_data, background_color, motif)

    # Generate response with enhanced settings
    raw_output = generate_attractive_response(attractive_prompt, model, tokenizer)
    attractive_lines = enforce_5_attractive_lines(raw_output, background_color, motif)

    print("\nGenerated ATTRACTIVE Ad Background Description:")
    for i, line in enumerate(attractive_lines, 1):
        print(f"{i}. {line}")

    # Create enhanced SD prompt
    sd_prompt = create_attractive_sd_prompt(attractive_lines, background_color, motif, appeal_factors)

    # Add appeal-specific enhancements to SD prompt
    enhancement_terms = []
    if appeal_factors['luxury_level'] == 'luxury':
        enhancement_terms.extend(['premium quality', 'luxury aesthetic', 'high-end'])
    if appeal_factors['target_emotion'] == 'exciting':
        enhancement_terms.extend(['dynamic', 'vibrant', 'energetic'])
    elif appeal_factors['target_emotion'] == 'sophisticated':
        enhancement_terms.extend(['elegant', 'refined', 'sophisticated'])

    if enhancement_terms:
        sd_prompt += ", " + ", ".join(enhancement_terms[:3])

    # Smart negative prompting - prevent boring results
    negative_prompt = create_smart_negative_prompt(product_data, motif)

    print(f"\nEnhanced SD Prompt: {sd_prompt}")
    print(f"Smart Negative Prompt: {negative_prompt}")

    # Free text model memory
    print("Freeing text model memory...")
    model.cpu()
    del model
    cleanup_memory()

    # Load Stable Diffusion with enhanced settings
    print("Loading Stable Diffusion pipeline...")
    pipe = StableDiffusionPipeline.from_pretrained(
        "runwayml/stable-diffusion-v1-5",
        torch_dtype=torch.float16
    ).to("cuda")

    # Generate attractive background with optimized parameters
    print("Generating attractive background image...")
    pipe.safety_checker = None

    background_image = pipe(
        prompt=sd_prompt,
        negative_prompt=negative_prompt,
        guidance_scale=9.0,  # Increased for stronger adherence to attractive elements
        num_inference_steps=60,  # Increased for higher quality
        height=768,  # Higher resolution for better quality
        width=768
    ).images[0]

    # Save and display
    background_image.save("attractive_background.png")
    print("Attractive background generated successfully!")

    plt.figure(figsize=(12, 10))
    plt.imshow(background_image)
    plt.axis("off")
    plt.title("Generated Attractive Ad Background", fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()

    return background_image, attractive_lines, sd_prompt
# ---------------- Memory Cleanup Function ----------------
def cleanup_memory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
# ---------------- Enhanced Examples ----------------
if __name__ == "__main__":
    # Example 1: Women's silver watch with attractive pink background
    print("=== EXAMPLE 1: Attractive Women's Watch Background ===")

    product_data = {
        "gender": "women",
        "baseColour": "silver",
        "usage": "casual",
        "subCategory": "watches",
        "masterCategory": "accessories",
        "year": 2016
    }

    bg1, lines1, prompt1 = generate_attractive_ad_background(
        product_data=product_data,
        background_color="soft pink with rose gold",
        motif="delicate flowers"
    )

    print("\n" + "="*70)

    # Example 2: Men's formal shoes with sophisticated background
    print("=== EXAMPLE 2: Sophisticated Men's Shoes Background ===")

    product_data_2 = {
        "gender": "men",
        "color": "black",
        "occasion": "formal",
        "type": "dress shoes",
        "style": "oxford",
        "material": "leather",
        "year": 2023
    }

    bg2, lines2, prompt2 = generate_attractive_ad_background(
        product_data=product_data_2,
        background_color="deep charcoal with gold accents",
        motif="geometric luxury"
    )

    print("\n" + "="*70)

    # Example 3: Trendy casual apparel with energetic background
    print("=== EXAMPLE 3: Energetic Casual Apparel Background ===")

    product_data_3 = {
        "gender": "unisex",
        "usage": "casual",
        "subCategory": "tshirts",
        "masterCategory": "apparel",
        "season": "summer",
        "year": 2024,
        "style": "trendy"
    }

    bg3, lines3, prompt3 = generate_attractive_ad_background(
        product_data=product_data_3,
        background_color="vibrant coral with electric blue",
        motif="dynamic waves"
    )

print("\n🎯 Enhanced version complete! Your backgrounds should now be much more attractive and customer-appealing!")

# ---------------- Pedestal Generation Functions ----------------
from PIL import Image, ImageDraw, ImageFilter, ImageEnhance
import numpy as np
import math

def create_cylindrical_pedestal(width, height, pedestal_color="white", shadow_intensity=0.3):
    """
    Creates a cylindrical pedestal with realistic 3D appearance
    """
    # Create base image
    img = Image.new('RGBA', (width, height), (0, 0, 0, 0))
    draw = ImageDraw.Draw(img)

    # Pedestal dimensions (relative to image size)
    pedestal_width = int(width * 0.25)  # 25% of image width
    pedestal_height = int(height * 0.15)  # 15% of image height

    # Center position
    center_x = width // 2
    center_y = int(height * 0.75)  # Position in lower part of image

    # Define pedestal colors based on input
    if pedestal_color.lower() == "white":
        base_color = (245, 245, 245)
        highlight_color = (255, 255, 255)
        shadow_color = (200, 200, 200)
    elif pedestal_color.lower() == "glass":
        base_color = (240, 248, 255, 180)  # Semi-transparent
        highlight_color = (255, 255, 255, 220)
        shadow_color = (180, 200, 220, 100)
    else:
        # Convert color name to RGB (basic implementation)
        base_color = (220, 220, 220)
        highlight_color = (255, 255, 255)
        shadow_color = (150, 150, 150)

    # Draw shadow (ellipse at the bottom)
    shadow_width = int(pedestal_width * 1.2)
    shadow_height = int(pedestal_width * 0.3)
    shadow_y = center_y + pedestal_height // 2

    # Create gradient shadow
    for i in range(10):
        alpha = int(shadow_intensity * 100 * (1 - i/10))
        shadow_expand = i * 2
        draw.ellipse([
            center_x - shadow_width//2 - shadow_expand,
            shadow_y - shadow_height//2 - i,
            center_x + shadow_width//2 + shadow_expand,
            shadow_y + shadow_height//2 + i
        ], fill=(0, 0, 0, alpha))

    # Draw pedestal cylinder (main body)
    # Side faces
    draw.rectangle([
        center_x - pedestal_width//2,
        center_y - pedestal_height//2,
        center_x + pedestal_width//2,
        center_y + pedestal_height//2
    ], fill=base_color)

    # Top ellipse (highlight)
    draw.ellipse([
        center_x - pedestal_width//2,
        center_y - pedestal_height//2 - 10,
        center_x + pedestal_width//2,
        center_y - pedestal_height//2 + 20
    ], fill=highlight_color)

    # Bottom ellipse
    draw.ellipse([
        center_x - pedestal_width//2,
        center_y + pedestal_height//2 - 15,
        center_x + pedestal_width//2,
        center_y + pedestal_height//2 + 15
    ], fill=shadow_color)

    return img

def create_geometric_pedestal(width, height, pedestal_color="white", shape="hexagon"):
    """
    Creates a geometric pedestal (hexagon, octagon, or square)
    """
    img = Image.new('RGBA', (width, height), (0, 0, 0, 0))
    draw = ImageDraw.Draw(img)

    # Pedestal dimensions
    pedestal_size = int(min(width, height) * 0.2)
    center_x = width // 2
    center_y = int(height * 0.75)

    # Colors
    if pedestal_color.lower() == "white":
        base_color = (245, 245, 245)
        highlight_color = (255, 255, 255)
        shadow_color = (200, 200, 200)
    else:
        base_color = (220, 220, 220)
        highlight_color = (255, 255, 255)
        shadow_color = (150, 150, 150)

    # Calculate polygon points
    if shape == "hexagon":
        sides = 6
    elif shape == "octagon":
        sides = 8
    else:  # square
        sides = 4

    # Generate polygon points
    points = []
    for i in range(sides):
        angle = 2 * math.pi * i / sides
        x = center_x + pedestal_size * math.cos(angle)
        y = center_y + pedestal_size * math.sin(angle)
        points.append((x, y))

    # Draw shadow
    shadow_points = [(x + 5, y + 5) for x, y in points]
    draw.polygon(shadow_points, fill=(0, 0, 0, 50))

    # Draw main shape
    draw.polygon(points, fill=base_color, outline=shadow_color, width=2)

    # Add highlight (smaller inner polygon)
    inner_points = []
    for i in range(sides):
        angle = 2 * math.pi * i / sides
        x = center_x + (pedestal_size * 0.7) * math.cos(angle)
        y = center_y + (pedestal_size * 0.7) * math.sin(angle)
        inner_points.append((x, y))

    draw.polygon(inner_points, fill=highlight_color)

    return img

def add_pedestal_to_background(background_image, pedestal_type="cylinder", pedestal_color="white", pedestal_style="modern"):
    """
    Main function to add pedestal to generated background
    """
    # Convert PIL Image if needed
    if hasattr(background_image, 'size'):
        bg_img = background_image.copy()
    else:
        bg_img = Image.fromarray(background_image)

    width, height = bg_img.size

    # Create pedestal based on type
    if pedestal_type == "cylinder":
        pedestal_img = create_cylindrical_pedestal(width, height, pedestal_color)
    elif pedestal_type == "geometric":
        shape = "hexagon" if pedestal_style == "modern" else "octagon"
        pedestal_img = create_geometric_pedestal(width, height, pedestal_color, shape)
    else:
        pedestal_img = create_cylindrical_pedestal(width, height, pedestal_color)

    # Apply blur for softer integration
    pedestal_img = pedestal_img.filter(ImageFilter.GaussianBlur(radius=0.5))

    # Composite pedestal onto background
    final_img = Image.alpha_composite(bg_img.convert('RGBA'), pedestal_img)

    return final_img.convert('RGB')

def generate_pedestal_placement_guide(background_with_pedestal):
    """
    Creates a visual guide showing where to place the product
    """
    img = background_with_pedestal.copy()
    draw = ImageDraw.Draw(img)

    width, height = img.size
    center_x = width // 2
    center_y = int(height * 0.75) - int(height * 0.15) // 2  # Top of pedestal

    # Draw placement guide (dashed circle)
    guide_radius = int(min(width, height) * 0.08)

    # Create dashed circle effect
    for angle in range(0, 360, 10):
        if angle % 20 == 0:  # Draw dash
            start_angle = math.radians(angle)
            end_angle = math.radians(angle + 8)

            x1 = center_x + guide_radius * math.cos(start_angle)
            y1 = center_y + guide_radius * math.sin(start_angle)
            x2 = center_x + guide_radius * math.cos(end_angle)
            y2 = center_y + guide_radius * math.sin(end_angle)

            draw.line([(x1, y1), (x2, y2)], fill=(255, 255, 255, 150), width=2)

    # Add center point
    draw.ellipse([
        center_x - 3, center_y - 3,
        center_x + 3, center_y + 3
    ], fill=(255, 255, 255, 180))

    return img

# ---------------- Enhanced Main Function with Pedestal ----------------
def generate_ad_background_with_pedestal(product_data, background_color, motif,
                                       pedestal_type="cylinder", pedestal_color="white",
                                       pedestal_style="modern", show_guide=True):
    """
    Enhanced version of the main function that includes pedestal generation
    """
    print("=== Starting Ad Background Generation with Pedestal ===")

    # Generate the background using your existing function

    background_image, ad_lines, sd_prompt = generate_attractive_ad_background(
    product_data, background_color, motif
)


    # Add pedestal to the background
    print(f"Adding {pedestal_type} pedestal in {pedestal_color}...")
    background_with_pedestal = add_pedestal_to_background(
        background_image, pedestal_type, pedestal_color, pedestal_style
    )

    # Save the result
    background_with_pedestal.save("background_with_solid_pedestal.png")
    print("Background with pedestal generated successfully!")

    # Display results
    fig, axes = plt.subplots(1, 3 if show_guide else 2, figsize=(15, 5))

    # Original background
    axes[0].imshow(background_image)
    axes[0].set_title("Original Background")
    axes[0].axis("off")

    # Background with pedestal
    axes[1].imshow(background_with_pedestal)
    axes[1].set_title("With Pedestal")
    axes[1].axis("off")

    # Placement guide (optional)
    if show_guide:
        guide_img = generate_pedestal_placement_guide(background_with_pedestal)
        axes[2].imshow(guide_img)
        axes[2].set_title("Product Placement Guide")
        axes[2].axis("off")
        guide_img.save("placement_guide.png")

    plt.tight_layout()
    plt.show()

    return background_with_pedestal, ad_lines, sd_prompt

# ---------------- Usage Examples ----------------
def example_usage():
    """
    Examples of how to use the enhanced function
    """
    # Your original product data
    product_data = {
        "gender": "women",
        "baseColour": "silver",
        "usage": "casual",
        "subCategory": "watches",
        "masterCategory": "accessories",
        "year": 2016
    }

    # Example 1: Cylinder pedestal (like your reference image)
    print("Example 1: Cylinder Pedestal")
    generate_ad_background_with_pedestal(
        product_data=product_data,
        background_color="turquoise",  # To match your reference image
        motif="geometric circles",
        pedestal_type="cylinder",
        pedestal_color="white",
        pedestal_style="modern"
    )

    # Example 2: Geometric pedestal
    print("\n" + "="*50)
    print("Example 2: Geometric Pedestal")
    generate_ad_background_with_pedestal(
        product_data=product_data,
        background_color="pink",
        motif="blue flowers",
        pedestal_type="geometric",
        pedestal_color="glass",
        pedestal_style="modern"
    )

# To integrate with your existing code, simply replace your final function call with:
# generate_ad_background_with_pedestal(product_data, background_color, motif)

if __name__ == "__main__":
    example_usage()
```

```python
from PIL import Image

def crop_to_content(img: Image.Image) -> Image.Image:
    """
    Crops an RGBA image to the bounding box of non-transparent pixels.
    """
    if img.mode != "RGBA":
        img = img.convert("RGBA")

    # Get alpha channel
    alpha = img.getchannel("A")

    # Get bounding box of non-transparent region
    bbox = alpha.getbbox()
    if bbox:
        img = img.crop(bbox)
    return img
```

```python
!pip install rembg



#version 4 - Multi-Color Platform Detection

from PIL import Image, ImageDraw, ImageStat, ImageFilter, ImageEnhance
import numpy as np
import cv2
import random
from rembg import remove
from IPython.display import display
import torch
from scipy import ndimage
from sklearn.cluster import KMeans
from scipy.spatial.distance import cdist
import json
import os

# Load platform metadata if available
metadata_path = "platform_metadata.json"
platform_metadata = None
if os.path.exists(metadata_path):
    with open(metadata_path, "r") as f:
        platform_metadata = json.load(f)
    print(f"📂 Loaded platform metadata: {platform_metadata}")

class PlatformSurfaceDetector:
    def __init__(self):
        self.model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)

    def detect_rendered_platforms(self, image):
        """Specialized detection for 3D rendered platforms - now supports multiple colors"""
        cv_image = cv2.cvtColor(np.array(image.convert('RGB')), cv2.COLOR_RGB2BGR)
        gray = cv2.cvtColor(cv_image, cv2.COLOR_BGR2GRAY)
        h, w = cv_image.shape[:2]

        print("🔍 Analyzing image for 3D rendered platforms...")

        platforms = []
        hsv = cv2.cvtColor(cv_image, cv2.COLOR_BGR2HSV)

        # MAIN FIX: Multiple color ranges for different platform types
        color_ranges = {
            'white': ([0, 0, 180], [180, 30, 255]),      # White/light gray (original)
            'green': ([35, 40, 100], [85, 255, 255]),    # Green platforms
            'blue': ([100, 50, 100], [130, 255, 255]),   # Blue platforms
            'brown': ([8, 50, 80], [25, 255, 200]),      # Wood/brown platforms
            'gray': ([0, 0, 60], [180, 50, 180]),        # Medium gray platforms
            'cream': ([15, 20, 150], [35, 80, 255]),     # Cream/beige platforms
            'pink': ([140, 30, 150], [170, 100, 255])    # Pink/rose platforms
        }

        # Method 1: Multi-color platform detection
        for color_name, (lower, upper) in color_ranges.items():
            lower = np.array(lower)
            upper = np.array(upper)
            color_mask = cv2.inRange(hsv, lower, upper)

            print(f"📊 {color_name.capitalize()} regions detected: {np.sum(color_mask > 0)} pixels")

            # Find contours in colored regions
            contours, _ = cv2.findContours(color_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

            for contour in contours:
                area = cv2.contourArea(contour)
                print(f"   {color_name} contour area: {area}")

                if area > 800:  # Same threshold
                    # Get bounding rectangle
                    x, y, w_rect, h_rect = cv2.boundingRect(contour)

                    # Check if it's roughly circular/elliptical (platform top)
                    aspect_ratio = w_rect / h_rect
                    if 0.7 < aspect_ratio < 1.4:  # Same aspect ratio check
                        # Try to fit a circle/ellipse
                        if len(contour) >= 5:
                            ellipse = cv2.fitEllipse(contour)
                            center = (int(ellipse[0][0]), int(ellipse[0][1]))
                            axes = (int(ellipse[1][0]/2), int(ellipse[1][1]/2))

                            platforms.append({
                                'center': center,
                                'axes': axes,
                                'bbox': (x, y, x + w_rect, y + h_rect),
                                'type': f'rendered_platform_{color_name}',
                                'area': area,
                                'contour': contour,
                                'color': color_name
                            })
                            print(f"✅ Found {color_name} platform at {center} with size {axes}")

        # Method 2: Edge-based circular detection (unchanged - works for any color)
        blurred = cv2.GaussianBlur(gray, (5, 5), 1)
        circles = cv2.HoughCircles(
            blurred,
            cv2.HOUGH_GRADIENT,
            dp=1,
            minDist=30,
            param1=30,
            param2=20,
            minRadius=15,
            maxRadius=min(w, h) // 3
        )

        if circles is not None:
            circles = np.round(circles[0, :]).astype("int")
            print(f"🔵 HoughCircles found {len(circles)} circles")

            for (x, y, r) in circles:
                # Use improved validation that works for any color
                if self.validate_rendered_platform_improved(cv_image, x, y, r):
                    platforms.append({
                        'center': (x, y),
                        'radius': r,
                        'bbox': (x-r, y-r, x+r, y+r),
                        'type': 'circular_platform',
                        'area': np.pi * r * r
                    })
                    print(f"✅ Validated circular platform at ({x}, {y}) radius {r}")

        # Method 3: Enhanced brightness patterns (works for any color)
        template_platforms = self.detect_by_brightness_patterns_improved(cv_image)
        platforms.extend(template_platforms)

        return platforms

    def validate_rendered_platform_improved(self, image, center_x, center_y, radius):
        """Improved validation that works for any color platform"""
        h, w = image.shape[:2]

        # Check bounds with padding
        if (center_x - radius < 0 or center_x + radius >= w or
            center_y - radius < 0 or center_y + radius >= h):
            return False

        # Extract region around the circle
        roi = image[max(0, center_y-radius):min(h, center_y+radius),
                   max(0, center_x-radius):min(w, center_x+radius)]

        if roi.size == 0:
            return False

        # IMPROVED: Check for uniform texture/color instead of just brightness
        roi_gray = cv2.cvtColor(roi, cv2.COLOR_BGR2GRAY)

        # Check uniformity (low variance = platform-like surface)
        variance = np.var(roi_gray)
        if variance < 2000:  # Uniform surface

            # Additional check: region should be distinct from surroundings
            # Sample surrounding area
            padding = radius // 2
            surround_roi = image[max(0, center_y-radius-padding):min(h, center_y+radius+padding),
                               max(0, center_x-radius-padding):min(w, center_x+radius+padding)]

            if surround_roi.size > 0:
                surround_gray = cv2.cvtColor(surround_roi, cv2.COLOR_BGR2GRAY)
                roi_mean = np.mean(roi_gray)
                surround_mean = np.mean(surround_gray)

                # Platform should be somewhat different from surroundings
                if abs(roi_mean - surround_mean) > 15:  # Some contrast
                    print(f"   Platform validated - variance: {variance:.1f}, contrast: {abs(roi_mean - surround_mean):.1f}")
                    return True

        return False

    def detect_by_brightness_patterns_improved(self, cv_image):
        """Improved brightness pattern detection for any color"""
        gray = cv2.cvtColor(cv_image, cv2.COLOR_BGR2GRAY)
        platforms = []

        # IMPROVEMENT: Use multiple brightness percentiles, not just brightest
        percentiles = [85, 75, 65, 35, 25, 15]  # Check bright, medium, and dark regions

        for percentile in percentiles:
            if percentile > 50:
                # Bright regions
                threshold = np.percentile(gray, percentile)
                mask = gray > threshold
                region_type = f'bright_{percentile}'
            else:
                # Dark regions (might be darker platforms)
                threshold = np.percentile(gray, percentile)
                mask = gray < threshold
                region_type = f'dark_{percentile}'

            # Find connected components
            num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(
                mask.astype(np.uint8), connectivity=8
            )

            for i in range(1, num_labels):  # Skip background (label 0)
                area = stats[i, cv2.CC_STAT_AREA]
                if area > 500:
                    x = stats[i, cv2.CC_STAT_LEFT]
                    y = stats[i, cv2.CC_STAT_TOP]
                    w = stats[i, cv2.CC_STAT_WIDTH]
                    h = stats[i, cv2.CC_STAT_HEIGHT]

                    # Check aspect ratio for platform-like shapes
                    aspect_ratio = w / h
                    if 0.5 < aspect_ratio < 2.0:
                        center_x = int(centroids[i][0])
                        center_y = int(centroids[i][1])

                        # Additional validation: check if region is uniform
                        roi = gray[y:y+h, x:x+w]
                        if roi.size > 0 and np.var(roi) < 1500:  # Uniform region
                            platforms.append({
                                'center': (center_x, center_y),
                                'bbox': (x, y, x + w, y + h),
                                'type': f'brightness_platform_{region_type}',
                                'area': area,
                                'aspect_ratio': aspect_ratio
                            })
                            print(f"✅ {region_type} platform found at ({center_x}, {center_y})")

        return platforms

    def find_best_platform_placement(self, platforms, product_size, image_size):
        """Find best platform with improved scoring for any color"""
        if not platforms:
            return None

        print(f"🎯 Evaluating {len(platforms)} platforms for placement...")

        best_platform = None
        best_score = -1

        img_width, img_height = image_size
        product_width, product_height = product_size

        for i, platform in enumerate(platforms):
            score = 0

            # Basic size compatibility - more lenient
            if platform['type'] == 'circular_platform':
                radius = platform.get('radius', 50)
                if product_width < radius * 2.5 and product_height < radius * 2.5:
                    score += 1.5
            else:
                x1, y1, x2, y2 = platform['bbox']
                platform_width = x2 - x1
                platform_height = y2 - y1

                if (product_width < platform_width * 1.2 and
                    product_height < platform_height * 1.2):
                    score += 1.5

            # Position scoring
            center_x, center_y = platform['center']

            # Prefer platforms in reasonable positions
            if 0.2 * img_height < center_y < 0.8 * img_height:
                score += 1.0

            # Size bonus
            area = platform.get('area', 0)
            if area > 1000:
                score += 0.5

            # IMPROVED: Type bonus for different detection methods
            platform_type = platform['type']
            if 'rendered_platform' in platform_type:
                score += 0.8  # Color-based detection is reliable
            elif platform_type == 'circular_platform':
                score += 0.9  # Circular detection is very reliable
            elif 'brightness_platform_bright' in platform_type:
                score += 0.7  # Bright regions are good
            elif 'brightness_platform_dark' in platform_type:
                score += 0.5  # Dark regions are less reliable but possible

            print(f"   Platform {i+1} ({platform['type']}): score = {score:.2f}")

            if score > best_score:
                best_score = score
                best_platform = platform

        print(f"🏆 Selected platform with score: {best_score:.2f}")
        return best_platform

    def calculate_product_position_on_platform(self, platform, product_size, image_size):
        """Calculate product position on platform"""
        product_width, product_height = product_size
        center_x, center_y = platform['center']

        # Center the product on the platform
        x = center_x - product_width // 2
        y = center_y - product_height // 2

        # Ensure within image bounds
        img_width, img_height = image_size
        x = max(0, min(x, img_width - product_width))
        y = max(0, min(y, img_height - product_height))

        return (int(x), int(y))

    def create_platform_aware_shadow(self, product_image, platform, product_position):
        """Create appropriate shadow for the platform"""
        shadow = Image.new("RGBA", product_image.size, (0, 0, 0, 0))
        mask = product_image.split()[3]

        # Shadow parameters based on platform type
        if platform['type'] == 'circular_platform':
            blur_radius = 15
            opacity = 0.25
            offset_x, offset_y = 4, 6
        elif 'rendered_platform' in platform['type']:
            blur_radius = 12
            opacity = 0.3
            offset_x, offset_y = 3, 5
        else:
            blur_radius = 10
            opacity = 0.35
            offset_x, offset_y = 5, 7

        # Create shadow
        shadow_pixels = shadow.load()
        mask_pixels = mask.load()

        for y in range(shadow.height):
            for x in range(shadow.width):
                if mask_pixels[x, y] > 0:
                    shadow_pixels[x, y] = (0, 0, 0, int(mask_pixels[x, y] * opacity))

        shadow = shadow.filter(ImageFilter.GaussianBlur(radius=blur_radius))
        return shadow, (offset_x, offset_y)

    def process_with_platform_detection(self, product_path, background_path, scale_factor=0.9):
        """Main processing function"""
        print("🚀 Starting platform detection process...")

        # Load product
        product_with_bg = Image.open(product_path).convert("RGB")
        product_no_bg = remove(product_with_bg)
        product_no_bg = crop_to_content(product_no_bg)


        # Resize product
        new_width = int(product_no_bg.width * scale_factor)
        new_height = int(product_no_bg.height * scale_factor)
        product_no_bg = product_no_bg.resize((new_width, new_height), Image.LANCZOS)
        print(f"📏 Product resized to: {new_width}x{new_height}")

        # Load background
        background_image = Image.open(background_path).convert("RGBA")
        print(f"🖼️ Background size: {background_image.size}")

        detected_platforms = self.detect_rendered_platforms(background_image)
        print(f"📊 Total platforms detected: {len(detected_platforms)}")

        best_platform = None
        product_position = None

        # If metadata is available, trust it for placement
        if platform_metadata:
            print("🎯 Using platform metadata for placement")
            top_surface = platform_metadata.get("top_surface", {})

            if "bbox" in top_surface:  # Cylindrical pedestal
                x1, y1, x2, y2 = top_surface["bbox"]
                surface_center = ((x1 + x2) // 2, (y1 + y2) // 2)

                best_platform = {
                    "center": surface_center,
                    "bbox": (x1, y1, x2, y2),
                    "type": platform_metadata.get("type", "metadata_platform"),
                    "area": top_surface.get("width", 0) * top_surface.get("height", 0)
                }

                # 🔑 Place product with its bottom touching pedestal top
                product_x = surface_center[0] - product_no_bg.width // 2
                product_y = y1 - product_no_bg.height
                product_position = (product_x, product_y)

            else:  # fallback if bbox missing
                best_platform = {
                    "center": tuple(platform_metadata["center"]),
                    "bbox": platform_metadata.get("bbox"),
                    "type": platform_metadata.get("type", "metadata_platform"),
                    "area": 0
                }

                # Default fallback placement (centered on platform center)
                cx, cy = best_platform["center"]
                product_x = cx - product_no_bg.width // 2
                product_y = cy - product_no_bg.height // 2
                product_position = (product_x, product_y)

        # If no metadata → fall back to detection
        if not best_platform and detected_platforms:
            best_platform = self.find_best_platform_placement(
                detected_platforms, product_no_bg.size, background_image.size
            )
            if best_platform:
                product_position = self.calculate_product_position_on_platform(
                    best_platform, product_no_bg.size, background_image.size
                )

        if best_platform:
            print(f"✅ Selected {best_platform['type']} platform at {best_platform['center']}")

            # Create shadow
            product_shadow, shadow_offset = self.create_platform_aware_shadow(
                product_no_bg, best_platform, product_position
            )

        else:
            print("⚠️ No suitable platform found, using center placement")
            product_position = (
                (background_image.width - product_no_bg.width) // 2,
                (background_image.height - product_no_bg.height) // 2
            )

            # Basic shadow
            shadow = Image.new("RGBA", product_no_bg.size, (0, 0, 0, 0))
            mask = product_no_bg.split()[3]
            shadow_pixels = shadow.load()
            mask_pixels = mask.load()
            for y in range(shadow.height):
                for x in range(shadow.width):
                    if mask_pixels[x, y] > 0:
                        shadow_pixels[x, y] = (0, 0, 0, int(mask_pixels[x, y] * 0.4))
            product_shadow = shadow.filter(ImageFilter.GaussianBlur(radius=10))
            shadow_offset = (5, 5)

        # Composite final image
        final_image = background_image.copy()

        # Add shadow
        shadow_position = (
            product_position[0] + shadow_offset[0],
            product_position[1] + shadow_offset[1]
        )
        final_image.paste(product_shadow, shadow_position, product_shadow)

        # Add product
        final_image.paste(product_no_bg, product_position, product_no_bg)

        print(f"🎨 Final placement at: {product_position}")

        return final_image, detected_platforms, best_platform


# Usage:
detector = PlatformSurfaceDetector()
final_image, detected_platforms, selected_platform = detector.process_with_platform_detection(
    "/content/watch3.jpg",
    "/content/background_with_solid_pedestal.png"
)

final_image.save("/content/platform_placement_result.png")
display(final_image)

if selected_platform:
    print(f"🎯 SUCCESS: Product placed on {selected_platform['type']} platform!")
else:
    print("❌ No platform detected - check detection parameters")
```
