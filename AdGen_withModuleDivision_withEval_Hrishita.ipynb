````python
import numpy as np
import colorsys
from sklearn.cluster import KMeans


class ColorHarmonyModule:
    """Color extraction + harmony context builder"""

    # -------------------------------------------------
    # EXISTING METHOD (UNCHANGED)
    # -------------------------------------------------
    @staticmethod
    def extract_dominant_colors(image, n_colors=5):
        img_array = np.array(image.convert('RGB'))
        pixels = img_array.reshape(-1, 3)

        if image.mode == 'RGBA':
            alpha = np.array(image.split()[3]).reshape(-1)
            pixels = pixels[alpha > 128]

        mask = np.sum(pixels > 240, axis=1) < 3
        pixels = pixels[mask]

        if len(pixels) < n_colors:
            return [(128, 128, 128)]

        kmeans = KMeans(n_clusters=n_colors, random_state=42, n_init=10)
        kmeans.fit(pixels)

        colors = kmeans.cluster_centers_.astype(int)
        labels = kmeans.labels_
        counts = [np.sum(labels == i) for i in range(n_colors)]
        order = np.argsort(counts)[::-1]

        return [tuple(colors[i]) for i in order]

    # -------------------------------------------------
    # 🆕 NEW: COLOR CONTEXT BUILDER (ALWAYS USED)
    # -------------------------------------------------
    @staticmethod
    def build_color_context(image, category):
        """
        Returns a STRING to be injected directly into the LLaMA prompt.
        """

        dominant_colors = ColorHarmonyModule.extract_dominant_colors(image)
        theory = ColorTheoryEngine()

        palettes = theory.get_harmonious_palettes(
            dominant_colors,
            rules=["complementary", "triadic", "analogous"]
        )

        # Limit for prompt sanity
        palettes = palettes[:6]

        # Format nicely for LLM consumption
        formatted = []
        for i, p in enumerate(palettes, 1):
            formatted.append(
                f"{i}. Palette: {p} | Use: background / accent / contrast"
            )

        return f"""
Color harmony analysis (computed, not guessed):

Dominant product colors:
{dominant_colors}

Harmonious palettes (based on color theory):
{chr(10).join(formatted)}

Guidance:
- Prefer palettes that enhance product visibility
- Backgrounds should not overpower the product
- Ensure strong text readability
"""
````

````python
class ColorTheoryEngine:

    def _to_unit(self, rgb):
        return tuple(c / 255.0 for c in rgb)

    def _to_rgb(self, rgb):
        return tuple(int(c * 255) for c in rgb)

    def get_complementary(self, rgb):
        h, s, v = colorsys.rgb_to_hsv(*self._to_unit(rgb))
        h = (h + 0.5) % 1.0
        return self._to_rgb(colorsys.hsv_to_rgb(h, s, v))

    def get_triadic(self, rgb):
        h, s, v = colorsys.rgb_to_hsv(*self._to_unit(rgb))
        return [
            self._to_rgb(colorsys.hsv_to_rgb((h + 0.333) % 1.0, s, v)),
            self._to_rgb(colorsys.hsv_to_rgb((h + 0.666) % 1.0, s, v))
        ]

    def get_analogous(self, rgb):
        h, s, v = colorsys.rgb_to_hsv(*self._to_unit(rgb))
        return [
            self._to_rgb(colorsys.hsv_to_rgb((h + 0.083) % 1.0, s, v)),
            self._to_rgb(colorsys.hsv_to_rgb((h - 0.083) % 1.0, s, v))
        ]

    def get_harmonious_palettes(self, colors, rules):
        palettes = []
        for rgb in colors:
            if "complementary" in rules:
                palettes.append([rgb, self.get_complementary(rgb)])
            if "triadic" in rules:
                palettes.append([rgb] + self.get_triadic(rgb))
            if "analogous" in rules:
                palettes.append([rgb] + self.get_analogous(rgb))
        return palettes
````

# Decluttering Module

````python
class DeclutteringModule:
    """
    IMPROVED MODULE: Background decluttering with adaptive complexity handling
    API-compatible with previous version
    """

    @staticmethod
    def analyze_background_complexity(background_image, threshold=0.6):
        """
        Analyze how cluttered/complex the background is
        Returns: complexity_score (0-1), high = cluttered
        """
        gray = np.array(background_image.convert('L')).astype(np.float32) / 255.0

        from scipy import ndimage

        # ---- 1. Multi-scale edge density ----
        edges_fine = np.hypot(
            ndimage.sobel(gray, axis=0),
            ndimage.sobel(gray, axis=1)
        )

        edges_coarse = np.hypot(
            ndimage.sobel(ndimage.gaussian_filter(gray, sigma=2), axis=0),
            ndimage.sobel(ndimage.gaussian_filter(gray, sigma=2), axis=1)
        )

        edge_density = (np.mean(edges_fine) * 0.7 + np.mean(edges_coarse) * 0.3)

        # ---- 2. Local texture variance (busy textures) ----
        local_variance = ndimage.generic_filter(
            gray, np.var, size=7
        )
        texture_complexity = np.mean(local_variance)

        # ---- 3. Global intensity variance ----
        global_variance = np.std(gray)

        # ---- Combined complexity score ----
        complexity = (
            edge_density * 0.45 +
            texture_complexity * 0.35 +
            global_variance * 0.20
        )

        return float(min(max(complexity, 0.0), 1.0))

    @staticmethod
    def apply_decluttering(background_image, intensity=0.5):
        """
        Apply adaptive decluttering effects
        """
        if intensity == 0:
            return background_image

        img = background_image.copy()
        gray = np.array(img.convert('L')).astype(np.float32) / 255.0

        from scipy import ndimage

        # ---- 1. Texture mask (where to declutter more) ----
        texture_map = ndimage.generic_filter(gray, np.var, size=7)
        texture_map = np.clip(texture_map / (texture_map.max() + 1e-6), 0, 1)

        # Convert mask to PIL
        mask = Image.fromarray((texture_map * 255).astype(np.uint8)).resize(
            img.size, Image.BILINEAR
        )

        # ---- 2. Two-level blur (selective) ----
        light_blur = img.filter(ImageFilter.GaussianBlur(radius=int(2 * intensity)))
        heavy_blur = img.filter(ImageFilter.GaussianBlur(radius=int(6 * intensity)))

        img = Image.composite(heavy_blur, light_blur, mask)

        # ---- 3. Saturation compression (non-linear) ----
        saturation = ImageEnhance.Color(img)
        saturation_factor = 1.0 - (intensity ** 1.3) * 0.45
        img = saturation.enhance(saturation_factor)

        # ---- 4. Contrast softening ----
        contrast = ImageEnhance.Contrast(img)
        contrast_factor = 1.0 - intensity * 0.30
        img = contrast.enhance(contrast_factor)

        # ---- 5. Slight brightness lift (perceptual calmness) ----
        brightness = ImageEnhance.Brightness(img)
        brightness_factor = 1.0 + intensity * 0.12
        img = brightness.enhance(brightness_factor)

        return img

    @staticmethod
    def suggest_declutter_intensity(complexity_score):
        """
        Suggest optimal decluttering intensity based on complexity
        """
        if complexity_score < 0.25:
            return 0.0, "Background is already clean"
        elif complexity_score < 0.45:
            return 0.25, "Light decluttering recommended"
        elif complexity_score < 0.65:
            return 0.5, "Moderate decluttering recommended"
        else:
            return 0.75, "Heavy decluttering recommended"
````

````python


# Create and display UI
print("\n" + "="*70)
print("Creating Interactive Interface...")
print("="*70 + "\n")

ui_controller = AdGeneratorUI()
ui_controller.display()

print("\n💡 TIP: Experiment with guidance scale and samplers for creative variety!")

#========================================================================#
#To extracr product colors

def extract_product_colors(image):
    """Extract dominant colors from product image"""
    try:
        # Resize image for faster processing
        img_small = image.resize((100, 100))
        img_array = np.array(img_small)

        # Reshape to list of RGB values
        pixels = img_array.reshape(-1, 3)

        # Remove fully transparent pixels if RGBA
        if img_array.shape[2] == 4:
            pixels = pixels[img_array.reshape(-1, 4)[:, 3] > 128]

        # Get 3 dominant colors using simple clustering
        from sklearn.cluster import KMeans
        kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
        kmeans.fit(pixels)

        colors = kmeans.cluster_centers_.astype(int)
        color_names = []

        for color in colors:
            # Convert RGB to color name approximation
            r, g, b = color
            color_names.append(f"RGB({r},{g},{b})")

        return color_names
    except:
        return ["Unable to extract"]

import webcolors

def get_color_name(hex_color):
    """Map hex to nearest color name or pass through if already a name."""
    if not isinstance(hex_color, str):
        return hex_color

    # If it’s already a color name like “Rose Pink”, return as is
    if not re.match(r"^#([A-Fa-f0-9]{6})$", hex_color.strip()):
        return hex_color

    try:
        return webcolors.hex_to_name(hex_color)
    except ValueError:
        # Find closest color name if exact match not found
        min_colors = {}
        rgb = webcolors.hex_to_rgb(hex_color)
        for key, name in webcolors.CSS3_HEX_TO_NAMES.items():
            r_c, g_c, b_c = webcolors.hex_to_rgb(key)
            rd = (r_c - rgb[0]) ** 2
            gd = (g_c - rgb[1]) ** 2
            bd = (b_c - rgb[2]) ** 2
            min_colors[(rd + gd + bd)] = name
        return min_colors[min(min_colors.keys())]



#======================================================================#

#new parsing function


def clean_and_extract_json2(raw_output: str) -> str:
    """
    Cleans and extracts valid JSON from the raw LLaMA output.
    Handles messy outputs like extra text, code fences, or malformed endings.
    Returns a JSON string.
    """
    print("Raw output is ---")
    print(raw_output)
    print()
    print(f"🧹 Cleaning model output (length: {len(raw_output)} chars)")

    if not raw_output or not isinstance(raw_output, str):
        raise ValueError("Empty or invalid model output.")

    # 1️⃣ Try to find JSON enclosed in ```json ... ```
    json_block_pattern = re.search(r"```json\s*(\{.*?\})\s*```", raw_output, re.DOTALL)
    if json_block_pattern:
        candidate = json_block_pattern.group(1).strip()
        try:
            json.loads(candidate)
            return candidate
        except json.JSONDecodeError:
            pass  # continue trying

    # 2️⃣ Try to find any {...} JSON-like block
    brace_pattern = re.search(r"\{.*\}", raw_output, re.DOTALL)
    if brace_pattern:
        candidate = brace_pattern.group(0).strip()

        # Remove unwanted trailing commas and fix common issues
        candidate = re.sub(r",\s*([\]}])", r"\1", candidate)
        candidate = candidate.replace("\n", " ").replace("\r", " ")
        candidate = re.sub(r"\s+", " ", candidate)

        try:
            json.loads(candidate)
            return candidate
        except json.JSONDecodeError as e:
            print(f"⚠️ JSON decode failed, attempting final cleanup: {e}")
            # Try to fix with loose matching
            candidate = candidate.rstrip('.,; ')
            try:
                json.loads(candidate)
                return candidate
            except Exception:
                pass

    # 3️⃣ Fallback — build minimal safe JSON if extraction fails
    print("⚠️ No valid JSON detected — using fallback.")
    return json.dumps({
        "suggested_colors": [],
        "suggested_motifs": []
    })






# For color suggestions

def generate_color_suggestions(product_desc, category, product_image=None):
    """
    Generate ONLY color suggestions (background colors + text colors)

    Args:
        product_desc: Product description text
        category: Product category key
        product_image: PIL Image of product (for color extraction)

    Returns:
        (background_colors, text_colors)
        - background_colors: list of color theme names
        - text_colors: list of dicts with {name, hex, reason}
    """
    print("\n🎨 Generating color recommendations...")

    # Extract dominant colors from product image
    #product_colors = []
    #if product_image is not None:
    #   chm = ColorHarmonyModule()
    #   product_colors = chm.extract_dominant_colors(product_image)

    color_context = ""
    if product_image:
          color_context = ColorHarmonyModule.build_color_context(
          product_image,
          category
      )

    #color_context = f"\nDominant product colors: {', '.join([f'RGB{c}' for c in product_colors])}" if product_colors else ""

    # Build prompt for LLaMA (ONLY asking for colors)
    prompt = f"""<|begin_of_text|><|start_header_id|>system<|end_header_id|>

You are an expert color theorist specializing in advertising design.

<|eot_id|><|start_header_id|>user<|end_header_id|>

Product: {product_desc}
Category: {CATEGORY_CONFIG[category]['name']}{color_context}

Suggest 4-5 background color themes that would create a harmonious, aesthetic advertisement background for this product. Consider color harmony, contrast, and visual appeal.Create a human-readable color description for an advertisement background.


Also suggest 3-4 text color options with explanations for readability.

Return your response in this exact JSON format:
{{
  "suggested_colors": ["Color Theme 1", "Color Theme 2", "Color Theme 3", "Color Theme 4", "Color Theme 5"],
  "suggested_text_colors": [
    {{"name": "White", "hex": "#FFFFFF", "reason": "High contrast on dark backgrounds"}},
    {{"name": "Gold", "hex": "#FFD700", "reason": "Luxury feel matching premium aesthetic"}},
    {{"name": "Navy", "hex": "#001F3F", "reason": "Professional and trustworthy"}}
  ]
}}

**IMPORTANT:**
- Return ONLY a JSON object
- Color themes should complement the product
- Be creative but professional

CRITICAL RULES:
1. Output ONLY the JSON object, nothing else
2. No explanations, no text before or after
3. Use proper JSON formatting
4. Ensure all strings are properly quoted
5. No trailing commas

<|eot_id|><|start_header_id|>assistant<|end_header_id|>

{{"""

    try:
        print("⏳ Creating LangChain pipeline from loaded LLaMA model...")

        # Check if LLaMA model is loaded
        if 'llama_model' not in globals() or 'tokenizer' not in globals():
            raise RuntimeError("❌ LLaMA model not found! Please run Step 0.5 first to load the model.")

        # Create pipeline for color suggestions
        color_pipeline = pipeline(
            "text-generation",
            model=llama_model,
            tokenizer=tokenizer,
            torch_dtype=torch.float16,
            device_map="auto",
            max_new_tokens=400,  # Enough for color suggestions
            temperature=0.7,
            top_p=0.9,
            do_sample=True,
            repetition_penalty=1.1,
            return_full_text=False,
            pad_token_id=tokenizer.eos_token_id,
        )

        print("✅ LangChain pipeline created!")
        print("🤖 Generating color suggestions...")

        # Generate
        result = color_pipeline(
            prompt,
            max_new_tokens=400,
            num_return_sequences=1,
            eos_token_id=tokenizer.eos_token_id,
            pad_token_id=tokenizer.eos_token_id,
        )

        # Extract generated text
        if result and len(result) > 0 and 'generated_text' in result[0]:
            raw_output = result[0]['generated_text']
        else:
            raw_output = ""

        if not raw_output or len(raw_output.strip()) == 0:
            print("⚠️ Model returned empty output. Using fallback suggestions.")
            default_config = CATEGORY_CONFIG[category]
            return default_config['colors'][:5], []

        # Complete the JSON if it was cut off
        if not raw_output.strip().endswith('}'):
            raw_output = raw_output + ']}'
        if not raw_output.strip().startswith('{'):
            raw_output = '{' + raw_output

        # Clean and parse JSON
        cleaned_json = clean_and_extract_json2(raw_output)
        suggestions = json.loads(cleaned_json)

        # Extract suggestions
        colors = suggestions.get('suggested_colors', [])
        text_colors = suggestions.get('suggested_text_colors', [])

        # Validate we got results
        if not colors:
            print("⚠️ No color suggestions generated. Using defaults.")
            default_config = CATEGORY_CONFIG[category]
            colors = default_config['colors'][:5]

        print(f"✅ Generated {len(colors)} background colors and {len(text_colors)} text colors!")

        # Clean up pipeline to free VRAM
        del color_pipeline
        torch.cuda.empty_cache()
        gc.collect()

        return colors, text_colors

    except Exception as e:
        print(f"⚠️ Could not generate color suggestions: {e}")
        import traceback
        traceback.print_exc()

        # Fallback to category defaults
        default_config = CATEGORY_CONFIG[category]
        print(f"   Using default {category} colors instead")
        return default_config['colors'][:5], []

````

````python



class MotifDesignValidator:
    """
    Validates motifs against design principles.
    This is the KEY component that makes it not just prompting.
    """

    def __init__(self):
        self.design_principles = self._load_design_principles()

    def _load_design_principles(self):
        """Design principles from established theory"""
        return {
            'specificity': {
                'weight': 0.3,
                'check': self._check_specificity,
                'description': 'Motifs should be specific, not generic'
            },
            'complexity': {
                'weight': 0.25,
                'check': self._check_complexity,
                'description': 'Visual complexity should be moderate'
            },
            'category_fit': {
                'weight': 0.25,
                'check': self._check_category_fit,
                'description': 'Motifs should match category conventions'
            },
            'coherence': {
                'weight': 0.2,
                'check': self._check_coherence,
                'description': 'Motifs should work together harmoniously'
            }
        }

    def validate(self, motifs, product_desc, category):
        """
        Validate motifs against design principles.
        Returns feedback for refinement.
        """
        scores = {}
        issues = []

        for name, principle in self.design_principles.items():
            score = principle['check'](motifs, product_desc, category)
            scores[name] = score

            if score < 0.6:
                issues.append(f"{name}_violation")

        overall_score = sum(
            scores[name] * principle['weight']
            for name, principle in self.design_principles.items()
        )

        return {
            'overall_score': overall_score,
            'principle_scores': scores,
            'issues': issues,
            'motifs': motifs,
            'category': category
        }

    # --------------------------------------------------
    # Design principle checks
    # --------------------------------------------------

    def _check_specificity(self, motifs, product_desc, category):
        """Check if motifs are specific enough"""
        generic_terms = [
            'modern', 'elegant', 'clean', 'simple', 'beautiful',
            'nice', 'good', 'professional', 'sleek', 'stylish'
        ]

        specificity_scores = []

        for motif in motifs:
            motif_lower = motif.lower()

            is_generic = any(term in motif_lower for term in generic_terms)

            has_color = any(color in motif_lower for color in [
                'red', 'blue', 'gold', 'silver', 'black', 'white',
                'green', 'purple', 'gradient', 'metallic'
            ])

            has_shape = any(shape in motif_lower for shape in [
                'circular', 'triangular', 'hexagonal', 'angular',
                'curved', 'linear', 'radial', 'grid'
            ])

            has_texture = any(tex in motif_lower for tex in [
                'texture', 'grain', 'fabric', 'silk',
                'glass', 'paper', 'wood', 'marble'
            ])

            if is_generic and not (has_color or has_shape or has_texture):
                score = 0.2
            elif has_color or has_shape or has_texture:
                score = 0.9
            else:
                score = 0.5

            specificity_scores.append(score)

        return np.mean(specificity_scores) if specificity_scores else 0.5

    def _check_complexity(self, motifs, product_desc, category):
        """Ensure visual complexity is moderate"""
        high_complexity_terms = [
            'intricate', 'detailed', 'complex', 'ornate',
            'elaborate', 'busy', 'dense', 'layered', 'fractal'
        ]

        scores = []

        for motif in motifs:
            motif_lower = motif.lower()
            word_count = len(motif.split())

            complexity_hits = sum(
                1 for term in high_complexity_terms if term in motif_lower
            )

            if complexity_hits >= 2 or word_count > 8:
                score = 0.3
            elif complexity_hits == 1 or word_count > 5:
                score = 0.6
            else:
                score = 0.9

            scores.append(score)

        return np.mean(scores) if scores else 0.5

    def _check_category_fit(self, motifs, product_desc, category):
        """Check if motifs match category conventions"""
        category_keywords = {
            'luxury': {
                'positive': ['gold', 'silk', 'elegant', 'marble', 'metallic', 'premium'],
                'negative': ['cartoon', 'playful', 'neon', 'busy']
            },
            'tech': {
                'positive': ['grid', 'geometric', 'angular', 'circuit', 'digital'],
                'negative': ['organic', 'floral', 'vintage', 'rustic']
            },
            'beauty': {
                'positive': ['soft', 'gradient', 'floral', 'pearl', 'pastel'],
                'negative': ['harsh', 'industrial', 'dark']
            },
            'fashion': {
                'positive': ['bold', 'pattern', 'texture', 'fabric'],
                'negative': ['clinical', 'sterile']
            },
            'food': {
                'positive': ['organic', 'warm', 'natural', 'rustic'],
                'negative': ['metallic', 'industrial', 'abstract']
            }
        }

        rules = category_keywords.get(category, category_keywords['luxury'])
        scores = []

        for motif in motifs:
            motif_lower = motif.lower()

            positive = sum(kw in motif_lower for kw in rules['positive'])
            negative = sum(kw in motif_lower for kw in rules['negative'])

            if negative > 0:
                score = 0.2
            elif positive >= 2:
                score = 1.0
            elif positive == 1:
                score = 0.7
            else:
                score = 0.5

            scores.append(score)

        return np.mean(scores) if scores else 0.5

    def _check_coherence(self, motifs, product_desc, category):
        """Check if motifs work together harmoniously"""
        if len(motifs) <= 1:
            return 1.0

        style_keywords = {
            'geometric': ['grid', 'angular', 'linear', 'hexagonal'],
            'organic': ['organic', 'floral', 'fluid', 'curved'],
            'minimal': ['minimal', 'clean', 'subtle'],
            'complex': ['intricate', 'ornate', 'layered']
        }

        style_counts = {k: 0 for k in style_keywords}

        for motif in motifs:
            motif_lower = motif.lower()
            for style, keywords in style_keywords.items():
                if any(kw in motif_lower for kw in keywords):
                    style_counts[style] += 1

        conflicts = 0
        if style_counts['geometric'] > 0 and style_counts['organic'] > 0:
            conflicts += 1
        if style_counts['minimal'] > 0 and style_counts['complex'] > 2:
            conflicts += 1

        if conflicts >= 2:
            return 0.3
        elif conflicts == 1:
            return 0.6
        else:
            return 0.9

def _llama_generate_motifs(
    llama_model,
    tokenizer,
    product_desc,
    category,
    category_config,
    creative=True,
    feedback=None
):
    """
    Internal function to generate motifs using LLaMA.

    Args:
        llama_model: The loaded LLaMA model
        tokenizer: The tokenizer
        product_desc: Product description
        category: Category key
        category_config: CATEGORY_CONFIG dict
        creative: If True, use higher temperature
        feedback: Optional validation feedback for refinement

    Returns:
        list of motif strings
    """
    try:
        temperature = 0.8 if creative else 0.6

        # Build appropriate prompt
        if feedback and feedback.get('issues'):
            prompt = build_refinement_prompt(
                motifs=feedback.get('motifs', []),
                issues=feedback['issues'],
                product_desc=product_desc,
                category=category
            )
        else:
            # Initial generation prompt
            prompt = f"""<|begin_of_text|><|start_header_id|>system<|end_header_id|>

You are an expert design consultant specializing in visual motifs and patterns for advertising.

<|eot_id|><|start_header_id|>user<|end_header_id|>

Product: {product_desc}
Category: {category_config[category]['name']}

Suggest 5 specific design motifs and visual patterns that would enhance this product's appeal in an advertisement background. Consider the product category and current aesthetic trends.

Be SPECIFIC - avoid generic terms. Include details about colors, shapes, or textures where appropriate.

Return your response in this exact JSON format:
{{
  "suggested_motifs": ["Specific Motif 1", "Specific Motif 2", "Specific Motif 3", "Specific Motif 4", "Specific Motif 5"]
}}

CRITICAL RULES:
1. Output ONLY the JSON object, nothing else
2. No explanations, no text before or after
3. Use proper JSON formatting
4. Ensure all strings are properly quoted
5. No trailing commas

<|eot_id|><|start_header_id|>assistant<|end_header_id|>

{{"""

        # Create pipeline
        pipe = pipeline(
            "text-generation",
            model=llama_model,
            tokenizer=tokenizer,
            torch_dtype=torch.float16,
            device_map="auto",
            max_new_tokens=350,
            temperature=temperature,
            top_p=0.9,
            do_sample=True,
            return_full_text=False,
            pad_token_id=tokenizer.eos_token_id,
        )

        # Generate
        out = pipe(prompt, num_return_sequences=1)
        raw = out[0]["generated_text"] if out else ""

        if not raw:
            return []

        # Parse motifs
        motifs = parse_motifs_from_json(raw)
        return motifs

    except Exception as e:
        print(f"   ⚠️ Error in _llama_generate_motifs: {e}")
        return []
    finally:
        # Cleanup
        try:
            if 'pipe' in locals():
                del pipe
            torch.cuda.empty_cache()
            gc.collect()
        except:
            pass



def parse_motifs_from_json(json_string):
    """
    Extract motifs from JSON response.
    Handles both clean and messy JSON outputs.
    """
    try:
        # AGGRESSIVE CLEANING FOR COMMON LLAMA ISSUES

        # 1. Look for array pattern first (most reliable)
        array_pattern = r'\[\s*"[^"]+"\s*(?:,\s*"[^"]+"\s*)*\]'
        array_match = re.search(array_pattern, json_string, re.DOTALL)

        if array_match:
            array_str = array_match.group(0)
            try:
                # Try to parse just the array
                motifs_list = json.loads(array_str)
                if motifs_list and isinstance(motifs_list, list):
                    print(f"   ✅ Extracted {len(motifs_list)} motifs from array pattern")
                    return [str(m) for m in motifs_list]
            except:
                pass

        # 2. Try to fix incomplete JSON (missing opening brace)
        json_string_stripped = json_string.strip()

        # If starts with a key like "suggested_motifs", add opening brace
        if not json_string_stripped.startswith('{') and '"' in json_string_stripped:
            if json_string_stripped.endswith('}'):
                json_string_fixed = '{' + json_string_stripped
            else:
                json_string_fixed = '{' + json_string_stripped + '}'

            try:
                data = json.loads(json_string_fixed)
                motifs = (
                    data.get('suggested_motifs') or
                    data.get('design_motifs') or
                    data.get('motifs') or
                    []
                )
                if motifs:
                    print(f"   ✅ Fixed incomplete JSON and extracted {len(motifs)} motifs")
                    return [str(m) for m in motifs]
            except:
                pass

        # 3. Try standard cleaning
        cleaned = clean_and_extract_json2(json_string)
        data = json.loads(cleaned)

        # Try different possible keys
        motifs = (
            data.get('suggested_motifs') or
            data.get('design_motifs') or
            data.get('motifs') or
            []
        )

        # Ensure all items are strings
        if motifs:
            print(f"   ✅ Extracted {len(motifs)} motifs via standard cleaning")
            return [str(m) if not isinstance(m, str) else m for m in motifs]

        return []

    except Exception as e:
        print(f"   ⚠️ Failed to parse motifs: {e}")
        return []


def build_refinement_prompt(motifs, issues, product_desc, category):
    """
    Build a prompt for refining motifs based on validation feedback.
    """
    # Map issues to human-readable feedback
    feedback_map = {
        'specificity_violation': 'Make motifs more specific and detailed (avoid generic terms like "modern" or "elegant")',
        'complexity_violation': 'Reduce visual complexity (avoid overly intricate or busy designs)',
        'category_fit_violation': f'Better align motifs with {category} category conventions',
        'coherence_violation': 'Ensure motifs work together harmoniously (avoid conflicting styles)'
    }

    feedback_text = "\n".join([f"- {feedback_map.get(issue, issue)}" for issue in issues])

    prompt = f"""<|begin_of_text|><|start_header_id|>system<|end_header_id|>

You are an expert design consultant. Refine these design motifs based on the feedback provided.

<|eot_id|><|start_header_id|>user<|end_header_id|>

Product: {product_desc}
Category: {category}

Previous motifs:
{json.dumps(motifs, indent=2)}

Issues to address:
{feedback_text}

Generate IMPROVED motifs that address these issues.

Return your response in this exact JSON format:
{{
  "suggested_motifs": ["Improved Motif 1", "Improved Motif 2", "Improved Motif 3", "Improved Motif 4", "Improved Motif 5"]
}}

CRITICAL RULES:
1. Output ONLY the JSON object, nothing else
2. No explanations, no text before or after
3. Use proper JSON formatting
4. Ensure all strings are properly quoted
5. No trailing commas

<|eot_id|><|start_header_id|>assistant<|end_header_id|>

{{"""

    return prompt





class MotifGenerationModule:
    def __init__(self, llama_model, tokenizer):
        self.llama_model = llama_model
        self.tokenizer = tokenizer
        self.enabled = True   # <-- IMPORTANT DEFAULT
        self.validator = MotifDesignValidator()
        self.max_iterations = 3

    def generate(self, product_desc, category):
        motifs_history = []
        validation_feedback = None

        for iteration in range(self.max_iterations):
            if iteration == 0:
                motifs = self._generate_initial(product_desc, category)
            else:
                motifs = self._refine(product_desc, category, validation_feedback)

            if not motifs:
                break

            motifs_history.append(motifs)
            validation_feedback = self.validator.validate(motifs, product_desc, category)

            if validation_feedback["overall_score"] >= 0.75:
                break

        return motifs_history[-1] if motifs_history else []

    # --- INTERNAL METHODS (shortened but equivalent) ---

    def _generate_initial(self, product_desc, category):
        return _llama_generate_motifs(
            self.llama_model, self.tokenizer, product_desc, category, creative=True
        )

    def _refine(self, product_desc, category, feedback):
        return _llama_generate_motifs(
            self.llama_model,
            self.tokenizer,
            product_desc,
            category,
            creative=False,
            feedback=feedback
        )


#Generating motif suggestions



# 4. MAIN MOTIF GENERATION FUNCTION (Completely Fixed)
# ============================================================================

def generate_motif_suggestions(product_desc, category):
    """
    Generate ONLY motif/design pattern suggestions with iterative refinement.

    Args:
        product_desc: Product description text
        category: Product category key

    Returns:
        list of motif/pattern names
    """
    print("\n✨ Generating motif recommendations...")

    # Initialize validator
    validator = MotifDesignValidator()
    max_iterations = 2  # Limit iterations to avoid excessive API calls

    try:
        # Check if LLaMA model is loaded
        if 'llama_model' not in globals() or 'tokenizer' not in globals():
            raise RuntimeError("❌ LLaMA model not found! Please run Step 0.5 first to load the model.")

        if 'CATEGORY_CONFIG' not in globals():
            raise RuntimeError("❌ CATEGORY_CONFIG not found!")

        # ===================================================================
        # ITERATION 1: Initial Generation
        # ===================================================================
        print("   🎨 Generating initial motifs...")

        initial_motifs = _llama_generate_motifs(
            llama_model=llama_model,
            tokenizer=tokenizer,
            product_desc=product_desc,
            category=category,
            category_config=CATEGORY_CONFIG,
            creative=True,
            feedback=None
        )

        if not initial_motifs:
            print("   ⚠️ No motifs generated. Using fallback.")
            default_config = CATEGORY_CONFIG[category]
            return default_config['motifs'][:5]

        print(f"   ✅ Generated {len(initial_motifs)} initial motifs")

        # ===================================================================
        # VALIDATION
        # ===================================================================
        print("   🔍 Validating motifs...")

        validation = validator.validate(
            motifs=initial_motifs,
            product_desc=product_desc,
            category=category
        )

        print(f"   📊 Validation score: {validation['overall_score']:.2f}")
        print(f"   📋 Principle scores: {validation['principle_scores']}")

        # If quality is good enough, return immediately
        if validation["overall_score"] >= 0.70:
            print("   ✅ Motifs passed validation!")
            return initial_motifs

        # ===================================================================
        # ITERATION 2: Refinement (if needed)
        # ===================================================================
        if validation['issues']:
            print(f"   ⚠️ Issues detected: {validation['issues']}")
            print("   🔄 Refining motifs...")

            refined_motifs = _llama_generate_motifs(
                llama_model=llama_model,
                tokenizer=tokenizer,
                product_desc=product_desc,
                category=category,
                category_config=CATEGORY_CONFIG,
                creative=False,  # Less creative for refinement
                feedback=validation
            )

            if refined_motifs:
                # Re-validate
                refined_validation = validator.validate(
                    motifs=refined_motifs,
                    product_desc=product_desc,
                    category=category
                )

                print(f"   📊 Refined score: {refined_validation['overall_score']:.2f}")

                # Use refined motifs if they're better
                if refined_validation['overall_score'] >= validation['overall_score']:
                    print("   ✅ Using refined motifs!")
                    return refined_motifs
                else:
                    print("   ℹ️ Refinement didn't improve quality, using initial motifs")
                    return initial_motifs
            else:
                print("   ⚠️ Refinement failed, using initial motifs")
                return initial_motifs

        return initial_motifs

    except Exception as e:
        print(f"   ⚠️ Could not generate motif suggestions: {e}")
        import traceback
        traceback.print_exc()

        # Fallback to category defaults
        try:
            default_config = CATEGORY_CONFIG[category]
            print(f"   ℹ️ Using default {category} motifs instead")
            return default_config['motifs'][:5]
        except:
            return ["Geometric Patterns", "Soft Gradients", "Minimal Design", "Clean Lines", "Professional"]













````

````python

# ======================================================================#
# CELL 4: LLaMA Pipeline Setup
# ============================================================================

def create_llama_pipeline():
    """Create LangChain pipeline using pre-loaded LLaMA model"""
    print("⏳ Creating LangChain pipeline from loaded LLaMA model...")

    if 'llama_model' not in globals() or 'tokenizer' not in globals():
        raise RuntimeError("❌ LLaMA model not found! Please run Step 0.5 first to load the model.")


    generation_pipeline = pipeline(
    "text-generation",
    model=llama_model,
    tokenizer=tokenizer,
    torch_dtype=torch.float16,
    device_map="auto",
    # max_new_tokens=1500,
    max_new_tokens=800,
    temperature=0.5,
    top_p=0.85,
    do_sample=True,
    repetition_penalty=1.15,
    return_full_text=False,
)

    llm = HuggingFacePipeline(pipeline=generation_pipeline)

    print("✅ LangChain pipeline created successfully!")
    return llm, generation_pipeline

# ============================================================================
# CELL 5: LangChain Components
# ============================================================================

from typing import Literal

class SceneArtistInstructions(BaseModel):
    prompt: str = Field(description="Detailed background prompt without the product")

class TypographerInstructions(BaseModel):
    headline: str = Field(description="Main headline (max 20 chars)")
    tagline: str = Field(description="Secondary tagline (max 30 chars)")
    font_prompt: Literal["Poppins-Bold.ttf", "Poppins-Regular.ttf", "Ubuntu-Regular.ttf", "Lato-Regular.ttf"]
    headline_size: int = Field(description="Font size 40-120px")
    tagline_size: int = Field(description="Font size 20-60px")
    color_prompt: str = Field(description="Hex color code")

class CreativeBrief(BaseModel):
    campaign_concept: str
    overall_mood: str
    layout_archetype: str
    scene_artist_instructions: SceneArtistInstructions
    typographer_instructions: TypographerInstructions
    compositor_instructions: str

parser = JsonOutputParser(pydantic_object=CreativeBrief)

def build_enhanced_prompt(user_config):
    """Build prompt with user selections"""
    category = user_config['category']
    category_info = CATEGORY_CONFIG[category]

    # Build the JSON example separately to avoid f-string conflicts
    json_example = {
        "campaign_concept": "brief catchy concept (5-10 words)",
        "overall_mood": "mood keywords (3-5 words)",
        "layout_archetype": user_config['composition'],
        "scene_artist_instructions": {
            "prompt": f"detailed background description with {user_config['color']} colors, {user_config['motif']} elements, {user_config['lighting']} lighting, professional commercial photography style"
        },
        "typographer_instructions": {
            "headline": "compelling headline (max 4 words)",
            "tagline": "supporting tagline (max 6 words)",
            "font_prompt": "Poppins-Bold.ttf",
            "headline_size": 80,
            "tagline_size": 40,
            "color_prompt": "#FFFFFF"
        },
        "compositor_instructions": "brief placement guidance (1 sentence)"
    }

    # Convert to formatted JSON string
    json_str = json.dumps(json_example, indent=2)

    template = f"""<|begin_of_text|><|start_header_id|>system<|end_header_id|>


You are an expert creative director AI for a high-end advertising agency. Your purpose is to conceptualize and generate a complete creative brief for a professional product advertisement.
Your primary goal is **visual harmony**. You must design a background scene that makes the user's provided product look like it truly belongs there. The final image should be a coherent, professional advertisement frame.

CRITICAL RULES:
1. Output ONLY the JSON object, nothing else
2. No explanations, no text before or after
3. Use proper JSON formatting
4. Ensure all strings are properly quoted
5. No trailing commas

<|eot_id|><|start_header_id|>user<|end_header_id|>

Product Details:
- Description: {user_config['product_description']}
- Category: {category_info['name']}
- Color Theme: {user_config['color']}
- Design Motif: {user_config['motif']}
- Lighting Style: {user_config['lighting']}
- Composition: {user_config['composition']}

Background Rules:
- No indoor environments such as bedrooms, living rooms, bathrooms, kitchens, or offices
- No landscapes, gardens, deserts, sceneries
- No nature scenes : No hills, grass, mountains, skies, sun, clouds, wildflowers, landscapes.
- No furniture or household items including sofas, beds, tables, chairs, lamps, or shelves
- Background must remain non-distracting to keep focus on the product

STRICT SEMANTIC CONSTRAINTS (MANDATORY):
- Do NOT use words or concepts related to:
  "serene", "peaceful", "beach", "sand", "dunes",
  "sky", "sun", "sunset", "horizon", "landscape", "outdoor", "scenery", "nature",
  "garden", "desert", "ocean", "sea", "clouds", "mountains"




Create a creative brief with this EXACT structure:
{json_str}

Output the JSON now:<|eot_id|><|start_header_id|>assistant<|end_header_id|>
"""

    return template




def clean_and_extract_json(raw_llm_output: str) -> str:
    """
    Aggressively cleans LLaMA's output to extract valid JSON.
    Handles multiple edge cases and malformed outputs.
    """
    print(f"\n   🔍 Cleaning output (length: {len(raw_llm_output)} chars)")

    # Strategy 1: Look for JSON in ```json``` code blocks
    json_block_pattern = r'```json\s*(\{.*?\})\s*```'
    json_block_match = re.search(json_block_pattern, raw_llm_output, re.DOTALL)
    if json_block_match:
        extracted = json_block_match.group(1).strip()
        print(f"   ✅ Found JSON in ```json``` block")
        return extracted

    # Strategy 2: Look for JSON in any ``` ``` code blocks
    any_block_pattern = r'```\s*(\{.*?\})\s*```'
    any_block_match = re.search(any_block_pattern, raw_llm_output, re.DOTALL)
    if any_block_match:
        extracted = any_block_match.group(1).strip()
        print(f"   ✅ Found JSON in generic code block")
        return extracted

    # Strategy 3: Use balanced brace counting to find the outermost JSON object
    first_brace = raw_llm_output.find('{')
    if first_brace == -1:
        print(f"   ❌ No opening brace found!")
        return raw_llm_output.strip()

    brace_count = 0
    end_pos = -1
    in_string = False
    escape_next = False

    for i in range(first_brace, len(raw_llm_output)):
        char = raw_llm_output[i]

        # Handle string literals (ignore braces inside strings)
        if escape_next:
            escape_next = False
            continue
        if char == '\\':
            escape_next = True
            continue
        if char == '"':
            in_string = not in_string
            continue
        if in_string:
            continue

        # Count braces outside of strings
        if char == '{':
            brace_count += 1
        elif char == '}':
            brace_count -= 1
            if brace_count == 0:
                end_pos = i + 1
                break

    # Fallback if unbalanced braces
    if end_pos == -1:
        print(f"   ⚠️ No matching closing brace found")
        # Instead of rfind (which might cut early), find the LAST closing brace that balances roughly
        last_brace = raw_llm_output.rstrip().rfind('}')
        if last_brace != -1 and last_brace > first_brace:
            extracted = raw_llm_output[first_brace:last_brace + 1]
            print(f"   ⚠️ Using last brace as fallback (pos {first_brace}–{last_brace})")
        else:
            extracted = raw_llm_output[first_brace:]
    else:
        extracted = raw_llm_output[first_brace:end_pos]
        print(f"   ✅ Extracted JSON from position {first_brace} to {end_pos}")

    # Clean common JSON issues
    extracted = re.sub(r',(\s*[}\]])', r'\1', extracted)


    # Validate and log errors clearly
    try:
        json.loads(extracted)

        print(f"   ✅ Extracted JSON is valid!")
        return extracted
    except json.JSONDecodeError as e:
        print(f"   ⚠️ Extracted JSON has errors: {e}")
        print(f"   📝 First 300 chars: {extracted[:300]}")
        print(f"   📝 Last 300 chars: {extracted[-300:]}")
        print(f"   ⚙️ Attempting to auto-fix trailing commas and unclosed braces...")

        # Auto-fix step: remove trailing commas and close braces if needed
        fixed = re.sub(r',(\s*[}\]])', r'\1', extracted)
        open_braces = fixed.count('{')
        close_braces = fixed.count('}')
        if open_braces > close_braces:
            fixed += '}' * (open_braces - close_braces)
        try:
            json.loads(fixed)
            print(f"   ✅ Auto-fixed JSON successfully!")
            return fixed
        except Exception as e2:
            print(f"   ❌ Still invalid after auto-fix: {e2}")
            return extracted




print("✅ LangChain components ready!")

# ============================================================================
# CELL 6: Stable Diffusion with Hyperparameters
# ============================================================================

sd_pipeline = None

def load_sd_pipeline():
    """Load Stable Diffusion pipeline"""
    global sd_pipeline

    if sd_pipeline is None:
        print("⏳ Loading Stable Diffusion pipeline...")
        pipe = StableDiffusionPipeline.from_pretrained(
            "SG161222/Realistic_Vision_V5.1_noVAE",
            torch_dtype=torch.float16,
            safety_checker=None,
            requires_safety_checker=False
        )
        pipe.to("cuda")
        sd_pipeline = pipe
        print("✅ Stable Diffusion loaded!")

    return sd_pipeline

def generate_background(prompt, category, quality_steps, config):
    """Generate background with enhanced hyperparameter control"""
    pipe = load_sd_pipeline()

    # Apply scheduler if different from default
    scheduler_name = config.get('scheduler', 'DPMSolverMultistepScheduler')
    if scheduler_name != pipe.scheduler.__class__.__name__:
        from diffusers import (
            DPMSolverMultistepScheduler,
            EulerAncestralDiscreteScheduler,
            EulerDiscreteScheduler,
            DDIMScheduler,
            LMSDiscreteScheduler
        )

        scheduler_map = {
            'DPMSolverMultistepScheduler': DPMSolverMultistepScheduler,
            'EulerAncestralDiscreteScheduler': EulerAncestralDiscreteScheduler,
            'EulerDiscreteScheduler': EulerDiscreteScheduler,
            'DDIMScheduler': DDIMScheduler,
            'LMSDiscreteScheduler': LMSDiscreteScheduler
        }

        scheduler_class = scheduler_map.get(scheduler_name, DPMSolverMultistepScheduler)
        #pipe.scheduler = scheduler_class.from_config(pipe.scheduler.config)
        pipe.scheduler = scheduler_class.from_config(
            pipe.scheduler.config,
            final_sigmas_type="sigma_min"
        )
        print(f"   🔄 Using scheduler: {scheduler_name}")

    category_suffix = CATEGORY_CONFIG[category]['prompt_suffix']
    force_no_humans = (
    "This is a product-only scene. No models are included."
)

    end = (
        "No human or human like forms. No women, men, children or similar forms"
    )

    enhanced_prompt = force_no_humans + prompt + category_suffix + end

    negative_prompt = ("text, letters, watermark, signature, blurry, low quality, "
                      "tables, lamps, amateur photography, poor lighting, cluttered, distracting elements,"
                      "human, woman, man, people, person, face, body, portrait, "
                      "model, skin, hair, selfie, hand, limbs, nude, flesh, ")

    # Get hyperparameters
    guidance_scale = config.get('guidance_scale', 7.5)
    seed = config.get('seed', -1)
    num_images = config.get('num_images', 1)
    cfg_rescale = config.get('cfg_rescale', 0.0)

    # Set seed for reproducibility
    generator = None
    if seed != -1:
        generator = torch.Generator(device="cuda").manual_seed(seed)
        print(f"   🎲 Using seed: {seed}")
    else:
        print(f"   🎲 Using random seed")

    print(f"   🎨 Guidance Scale: {guidance_scale}")
    print(f"   🖼️ Generating {num_images} image(s)...")

    # Generate images
    kwargs = {
        'prompt': enhanced_prompt,
        'negative_prompt': negative_prompt,
        'height': 1024,
        'width': 1024,
        'num_inference_steps': quality_steps,
        'guidance_scale': guidance_scale,
        'num_images_per_prompt': num_images,
    }

    if generator:
        kwargs['generator'] = generator

    # Add CFG rescale if supported and non-zero
    if cfg_rescale > 0:
        try:
            kwargs['guidance_rescale'] = cfg_rescale
            print(f"   🔬 CFG Rescale: {cfg_rescale}")
        except:
            print(f"   ⚠️ CFG Rescale not supported by this model version")

    results = pipe(**kwargs).images

    return results  # Returns list of images

print("✅ Stable Diffusion functions ready!")

# ============================================================================
# CELL 7: Enhanced Typography with Text Highlighting
# ============================================================================

def draw_text_with_highlight(draw, text, position, font, text_color, highlight_preset_key, canvas_width):
    """Draw text with optional highlight background"""
    highlight_config = TEXT_HIGHLIGHT_PRESETS[highlight_preset_key]

    if not highlight_config['enabled']:
        # No highlight - just draw text with shadow
        shadow_offset = 3
        fill_color = text_color + (255,) if len(text_color) == 3 else text_color
        draw.text((position[0] + shadow_offset, position[1] + shadow_offset),
                 text, font=font, fill=(0, 0, 0, 180))
        draw.text(position, text, font=font, fill=fill_color)
        return

    # Get text dimensions
    bbox = draw.textbbox(position, text, font=font)
    text_width = bbox[2] - bbox[0]
    text_height = bbox[3] - bbox[1]

    # Calculate highlight bar dimensions
    padding = highlight_config['padding']
    bar_x1 = 0  # Full width bar
    bar_x2 = canvas_width
    bar_y1 = position[1] - padding
    bar_y2 = position[1] + text_height + padding

    # Draw highlight background
    bar_color = highlight_config['bar_color']
    bar_opacity = int(highlight_config['bar_opacity'] * 255)

    highlight_image = Image.new("RGBA", (canvas_width, bar_y2 - bar_y1), (0,0,0,0))
    highlight_draw = ImageDraw.Draw(highlight_image)

    if highlight_config.get('gradient', False):
        # Draw gradient highlight
        for i in range(int(bar_y2 - bar_y1)):
            alpha = int(bar_opacity * (1 - i / (bar_y2 - bar_y1)))
            highlight_draw.line(
                [(0, i), (canvas_width, i)],
                fill=bar_color + (alpha,)
            )
    else:
        # Draw solid highlight
        highlight_draw.rectangle(
            [0, 0, canvas_width, bar_y2 - bar_y1],
            fill=bar_color + (bar_opacity,)
        )

    # Paste highlight image onto the main canvas
    draw._image.paste(highlight_image, (bar_x1, int(bar_y1)), highlight_image)

    # Draw text
    fill_color = text_color + (255,) if len(text_color) == 3 else text_color
    draw.text(position, text, font=font, fill=fill_color)

def render_typography_enhanced(instructions, width, height, config):
    """Enhanced text rendering with separate fonts and highlight controls"""
    # Extract text - USE CUSTOM TEXT IF PROVIDED
    headline_text = config.get('custom_headline') or instructions.get("headline", "Headline")
    tagline_text = config.get('custom_tagline') or instructions.get("tagline", "Tagline")

    # USE USER-SPECIFIED FONT SIZES
    headline_size = config.get('headline_size', 80)
    tagline_size = config.get('tagline_size', 40)

    print(f"   🔍 DEBUG - Headline size from config: {headline_size}")
    print(f"   🔍 DEBUG - Tagline size from config: {tagline_size}")

    # UPDATED: Use user-selected color with fallback to AI suggestion
    text_color_hex = config.get('text_color')  # Get user selection
    if not text_color_hex:  # Fallback to AI suggestion if not set
        color_prompt = instructions.get("color_prompt", "#FFFFFF")
        match = re.search(r'#([A-Fa-f0-9]{6})', color_prompt)
        text_color_hex = match.group(0) if match else "#FFFFFF"

    # Convert hex to RGB
    text_color = tuple(int(text_color_hex[i:i+2], 16) for i in (1, 3, 5))
    print(f"   🎨 Using text color: {text_color_hex}")

    # Get positions
    headline_y = TEXT_POSITION_OPTIONS[config['headline_position']]['y']
    tagline_y = TEXT_POSITION_OPTIONS[config['tagline_position']]['y']
    alignment = config['text_alignment']

    # Create canvas
    canvas = Image.new("RGBA", (width, height), (0, 0, 0, 0))
    draw = ImageDraw.Draw(canvas)

    # Load headline font with fallback
    headline_font = None
    headline_font_name = config.get('headline_font', 'Poppins-Bold.ttf')

    try:
        headline_font = ImageFont.truetype(headline_font_name, headline_size)
        print(f"   ✅ Loaded headline font: {headline_font_name} at {headline_size}px")
    except Exception as e:
        print(f"   ⚠️ Could not load {headline_font_name}: {e}")

        fallback_fonts = [
            '/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf',
            '/usr/share/fonts/truetype/liberation/LiberationSans-Bold.ttf',
            '/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf',
            'arial.ttf',
            'Arial.ttf'
        ]

        for fallback in fallback_fonts:
            try:
                headline_font = ImageFont.truetype(fallback, headline_size)
                print(f"   ✅ Using fallback font: {fallback} at {headline_size}px")
                break
            except:
                continue

    if headline_font is None:
        print(f"   ❌ WARNING: Using default font for headline")
        headline_font = ImageFont.load_default()

    # Load tagline font with fallback
    tagline_font = None
    tagline_font_name = config.get('tagline_font', 'Poppins-Regular.ttf')

    try:
        tagline_font = ImageFont.truetype(tagline_font_name, tagline_size)
        print(f"   ✅ Loaded tagline font: {tagline_font_name} at {tagline_size}px")
    except Exception as e:
        print(f"   ⚠️ Could not load {tagline_font_name}: {e}")

        fallback_fonts = [
            '/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf',
            '/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf',
            '/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf',
            'arial.ttf',
            'Arial.ttf'
        ]

        for fallback in fallback_fonts:
            try:
                tagline_font = ImageFont.truetype(fallback, tagline_size)
                print(f"   ✅ Using fallback font: {fallback} at {tagline_size}px")
                break
            except:
                continue

    if tagline_font is None:
        print(f"   ❌ WARNING: Using default font for tagline")
        tagline_font = ImageFont.load_default()

    # Calculate positions based on alignment
    margin = 60

    # HEADLINE
    h_bbox = draw.textbbox((0, 0), headline_text, font=headline_font)
    h_width = h_bbox[2] - h_bbox[0]

    if alignment == 'center':
        h_x = (width - h_width) / 2
    elif alignment == 'left':
        h_x = margin
    elif alignment == 'right':
        h_x = width - h_width - margin

    headline_pos = (h_x, headline_y)

    # TAGLINE
    t_bbox = draw.textbbox((0, 0), tagline_text, font=tagline_font)
    t_width = t_bbox[2] - t_bbox[0]

    if alignment == 'center':
        t_x = (width - t_width) / 2
    elif alignment == 'left':
        t_x = margin
    elif alignment == 'right':
        t_x = width - t_width - margin

    tagline_pos = (t_x, tagline_y)

    # Draw headline with highlight
    draw_text_with_highlight(
        draw,
        headline_text,
        headline_pos,
        headline_font,
        text_color,
        config['headline_highlight'],
        width
    )

    # Draw tagline with highlight
    draw_text_with_highlight(
        draw,
        tagline_text,
        tagline_pos,
        tagline_font,
        text_color,
        config['tagline_highlight'],
        width
    )

    # Debug info
    text_source = "Custom" if config.get('custom_headline') else "AI-Generated"
    print(f"   📝 Headline ({text_source}): '{headline_text}' - Font: {headline_font_name} - Size: {headline_size}px")
    print(f"   📝 Headline Position: {TEXT_POSITION_OPTIONS[config['headline_position']]['name']} (Y={headline_y})")
    print(f"   🎨 Headline Highlight: {TEXT_HIGHLIGHT_PRESETS[config['headline_highlight']]['name']}")
    text_source = "Custom" if config.get('custom_tagline') else "AI-Generated"
    print(f"   📝 Tagline ({text_source}): '{tagline_text}' - Font: {tagline_font_name} - Size: {tagline_size}px")
    print(f"   📝 Tagline Position: {TEXT_POSITION_OPTIONS[config['tagline_position']]['name']} (Y={tagline_y})")
    print(f"   🎨 Tagline Highlight: {TEXT_HIGHLIGHT_PRESETS[config['tagline_highlight']]['name']}")
    print(f"   📍 Alignment: {alignment.title()}")

    return canvas

print("✅ Enhanced typography functions ready!")

# ============================================================================
# CELL 8: Composition Function
# ============================================================================

def compose_final_ad(bg_img, product_img, text_layer, composition_style):
    """Compose final advertisement with better layering"""
    bg_w, bg_h = bg_img.size

    # Adjust product scale based on composition
    scale = 0.65 if "Split" in composition_style else 0.75
    new_w = int(bg_w * scale)
    aspect = product_img.height / product_img.width
    new_h = int(new_w * aspect)
    product_resized = product_img.resize((new_w, new_h), Image.LANCZOS)

    # Create shadow
    shadow = Image.new("RGBA", product_resized.size, (0, 0, 0, 0))
    shadow_layer = Image.new("RGBA", product_resized.size, (0, 0, 0, 180))
    shadow.paste(shadow_layer, (0, 0), product_resized.split()[3])
    shadow = shadow.filter(ImageFilter.GaussianBlur(30))

    final = bg_img.convert("RGBA")

    # Smart product placement (centered, bottom-aligned)
    bottom_margin = int(bg_h * 0.12)
    product_x = (bg_w - new_w) // 2
    product_y = max(0, min(bg_h - new_h - bottom_margin, bg_h - new_h))

    # Layer composition
    final.paste(shadow, (product_x, product_y + 50), shadow)
    final.paste(product_resized, (product_x, product_y), product_resized)

    # Overlay text
    final.paste(text_layer, (0, 0), text_layer)

    return final

print("✅ Composition functions ready!")

# ============================================================================
# CELL 9: Main Generation Function
# ============================================================================


def generate_advertisement():
    """
    Main generation function
    🔄 MODIFIED: Added decluttering module integration
    """
    # Get user configuration
    config = ui_controller.get_config()

    if config['product_image'] is None:
        print("❌ Please upload a product image first!")
        return

    if not config['product_description']:
        print("❌ Please provide a product description!")
        return

    print("\n" + "="*70)
    print("🚀 STARTING ADVERTISEMENT GENERATION")
    print("="*70)
    print(f"\n📦 Product: {config['product_description'][:50]}...")
    print(f"🎨 Category: {CATEGORY_CONFIG[config['category']]['name']}")
    print(f"🎨 Color: {config['color']}")
    print(f"✨ Motif: {config['motif']}")
    print(f"💡 Lighting: {config['lighting']}")
    print(f"📐 Layout: {config['composition']}")

    # 🔄 NEW: Show decluttering status
    if config['use_decluttering']:
        print(f"🧹 Decluttering: ✅ Enabled (Intensity: {config['declutter_intensity']:.1f})")
    else:
        print(f"🧹 Decluttering: ❌ Disabled")

    print(f"\n🔧 SD Parameters:")
    print(f"   Guidance Scale: {config['guidance_scale']}")
    print(f"   Seed: {config['seed']}")
    print(f"   Num Images: {config['num_images']}")
    print(f"   Scheduler: {config['scheduler']}")

    try:
        # Step 1: Create LangChain pipeline
        print("\n1️⃣ Creating LangChain pipeline from loaded LLaMA model...")
        llm, gen_pipeline = create_llama_pipeline()

        # Step 2: Generate creative brief with LLaMA
        print("2️⃣ Generating creative brief...")
        prompt_text = build_enhanced_prompt(config)
        print("   📤 Sending prompt to LLaMA...")

        raw_output = llm.invoke(prompt_text)
        print("\n" + "="*70)
        print("📝 RAW LLAMA OUTPUT:")
        print("="*70)
        print(raw_output[:500] + "..." if len(raw_output) > 500 else raw_output)
        print("="*70 + "\n")

        cleaned_json = clean_and_extract_json(raw_output)
        print("   🧹 Cleaned JSON:")
        print(cleaned_json[:500] + "..." if len(cleaned_json) > 500 else cleaned_json)
        print()

        try:
            creative_brief = json.loads(cleaned_json)
            print("✅ Creative brief generated from AI!")
            print("\n📋 Creative Brief:")
            print(json.dumps(creative_brief, indent=2)[:500] + "...")

            # Override text fields if user provided custom ones
            if config['custom_headline']:
                print(f"\n   📝 Overriding headline with custom: '{config['custom_headline']}'")
                creative_brief['typographer_instructions']['headline'] = config['custom_headline']

            if config['custom_tagline']:
                print(f"   📝 Overriding tagline with custom: '{config['custom_tagline']}'")
                creative_brief['typographer_instructions']['tagline'] = config['custom_tagline']

            print("   ✅ Using LLaMA's background description for Stable Diffusion")

        except json.JSONDecodeError as e:
            print(f"⚠️ Failed to parse AI creative brief: {e}")
            print(f"   Error at position: {e.pos}")
            print("\n   Using fallback creative brief...")

            creative_brief = {
                'campaign_concept': 'Premium Product Showcase',
                'overall_mood': 'Professional and Attractive',
                'layout_archetype': config['composition'],
                'scene_artist_instructions': {
                    'prompt': f"Professional advertising background with {config['color']} color scheme and {config['motif']} design elements, {config['lighting']} lighting, {config['composition']} composition"
                },
                'typographer_instructions': {
                    'headline': config.get('custom_headline') or 'Premium Quality',
                    'tagline': config.get('custom_tagline') or config['product_description'][:30],
                    'font_prompt': 'Poppins-Bold.ttf',
                    'headline_size': 80,
                    'tagline_size': 40,
                    'color_prompt': '#FFFFFF'
                }
            }

        print(f"   Campaign: {creative_brief.get('campaign_concept', 'N/A')}")
        print(f"   Headline: {creative_brief['typographer_instructions'].get('headline', 'N/A')}")
        print(f"   Tagline: {creative_brief['typographer_instructions'].get('tagline', 'N/A')}")

        # Step 3: Unload LLaMA to free VRAM
        print("\n3️⃣ Unloading LLaMA pipeline to free VRAM...")
        del llm, gen_pipeline
        torch.cuda.empty_cache()
        gc.collect()
        print("✅ VRAM cleared!")

        # Step 4: Generate background
        print("\n4️⃣ Generating background image(s)...")
        quality_map = {
            'Standard (25 steps)': 25,
            'High Quality (50 steps)': 50,
            'Ultra Quality (75 steps)': 75
        }
        steps = quality_map[config['quality']]

        bg_prompt = creative_brief['scene_artist_instructions']['prompt']
        background_images = generate_background(bg_prompt, config['category'], steps, config)
        print("✅ Background(s) generated!")

        # Display all generated backgrounds
        num_images = len(background_images)
        if num_images > 1:
            print(f"\n🖼️ Generated {num_images} variations. Displaying all:")
            fig, axes = plt.subplots(1, num_images, figsize=(6*num_images, 6))
            if num_images == 1:
                axes = [axes]
            for idx, bg_img in enumerate(background_images):
                axes[idx].imshow(bg_img)
                axes[idx].set_title(f"Variation {idx+1}")
                axes[idx].axis("off")
            plt.tight_layout()
            plt.show()

            print("\n📌 Using first variation for final composition")
            background = background_images[0]
        else:
            background = background_images[0]
            plt.figure(figsize=(10, 10))
            plt.imshow(background)
            plt.title("Generated Background")
            plt.axis("off")
            plt.show()

        # 🔄 CHANGE 10: DECLUTTERING MODULE INTEGRATION
        # =====================================================================
        # NEW STEP: Apply decluttering if enabled
        # =====================================================================
        original_background = background.copy()  # Keep original for comparison

        if config['use_decluttering']:
            print(f"\n🧹 APPLYING DECLUTTERING MODULE")
            print("="*70)
            print(f"   Intensity: {config['declutter_intensity']:.1f}")

            # Initialize decluttering module
            declutter_module = DeclutteringModule()

            # Analyze background complexity
            complexity = declutter_module.analyze_background_complexity(background)
            print(f"   📊 Background complexity score: {complexity:.3f}")

            # Get suggested intensity
            suggested_intensity, reason = declutter_module.suggest_declutter_intensity(complexity)
            print(f"   💡 AI suggestion: {reason} (Intensity: {suggested_intensity:.1f})")
            print(f"   👤 User selected: {config['declutter_intensity']:.1f}")

            # Apply decluttering
            print(f"   ⏳ Applying decluttering...")
            background = declutter_module.apply_decluttering(
                background,
                intensity=config['declutter_intensity']
            )
            print(f"   ✅ Decluttering applied!")

            # Show before/after comparison
            print(f"\n   📊 Showing before/after comparison...")
            fig, axes = plt.subplots(1, 2, figsize=(16, 8))

            axes[0].imshow(original_background)
            axes[0].set_title("Original Background", fontsize=14, fontweight='bold')
            axes[0].axis("off")

            axes[1].imshow(background)
            axes[1].set_title(f"Decluttered (Intensity: {config['declutter_intensity']:.1f})",
                             fontsize=14, fontweight='bold')
            axes[1].axis("off")

            plt.suptitle('🧹 Decluttering Module - Before & After',
                        fontsize=16, fontweight='bold', y=0.98)
            plt.tight_layout()
            plt.show()

            print("="*70)
        else:
            print(f"\n🧹 Decluttering: Skipped (disabled by user)")

        # =====================================================================
        # End of decluttering section
        # =====================================================================

        # Step 5: Create typography
        print("\n5️⃣ Creating typography with custom fonts and highlights...")
        text_layer = render_typography_enhanced(
            creative_brief['typographer_instructions'],
            background.width,
            background.height,
            config
        )
        print("✅ Typography created!")

        # Step 6: Compose final ad
        print("\n6️⃣ Composing final advertisement...")
        final_ad = compose_final_ad(
            background,
            config['product_image'],
            text_layer,
            config['composition']
        )
        print("✅ Advertisement complete!")

        # Display result
        print("\n" + "="*70)
        print("🎉 ADVERTISEMENT GENERATED SUCCESSFULLY!")
        print("="*70 + "\n")

        plt.figure(figsize=(12, 12))
        plt.imshow(final_ad)
        title = f"Final Advertisement - {CATEGORY_CONFIG[config['category']]['name']}"
        if config['use_decluttering']:
            title += f" (Decluttered: {config['declutter_intensity']:.1f})"
        plt.title(title, fontsize=16, fontweight='bold')
        plt.axis("off")
        plt.tight_layout()
        plt.show()

        # Save
        final_ad.save("generated_advertisement.png", quality=95, optimize=True)
        print("💾 Saved as 'generated_advertisement.png'")

        # 🔄 NEW: Save decluttering comparison if used
        if config['use_decluttering']:
            # Save original version too
            text_layer_orig = render_typography_enhanced(
                creative_brief['typographer_instructions'],
                original_background.width,
                original_background.height,
                config
            )
            final_ad_orig = compose_final_ad(
                original_background,
                config['product_image'],
                text_layer_orig,
                config['composition']
            )
            final_ad_orig.save("generated_advertisement_no_declutter.png", quality=95)
            print("💾 Saved original (no declutter) as 'generated_advertisement_no_declutter.png'")

        # If multiple images were generated, process all variations
        if num_images > 1:
            print(f"\n🎨 Creating complete ads for all {num_images} variations...")
            for idx, bg_img in enumerate(background_images):
                # Apply decluttering to each variation if enabled
                if config['use_decluttering']:
                    bg_img = declutter_module.apply_decluttering(
                        bg_img,
                        config['declutter_intensity']
                    )

                text_layer_var = render_typography_enhanced(
                    creative_brief['typographer_instructions'],
                    bg_img.width,
                    bg_img.height,
                    config
                )
                final_ad_var = compose_final_ad(
                    bg_img,
                    config['product_image'],
                    text_layer_var,
                    config['composition']
                )
                final_ad_var.save(f"generated_advertisement_variation_{idx+1}.png", quality=95)
                print(f"   💾 Saved variation {idx+1}")
            print(f"✅ All {num_images} variations saved!")

    except Exception as e:
        print(f"\n❌ Error during generation: {e}")
        import traceback
        traceback.print_exc()


    #======EVALUATION SECTION===========#

    from IPython.display import display, clear_output
    from ipywidgets import widgets, Layout

    print("\n" + "="*70)
    print("📝 EVALUATION SETUP")
    print("="*70)
    print("Did you use AI Recommendations for this ad?")

    # Create dropdown widget
    method_dropdown = widgets.Dropdown(
        options=[
            ('❌ NO - Used category defaults (without AI reco)', 'without_ai_reco'),
            ('✅ YES - Clicked AI Recommendations button (with AI reco)', 'with_ai_reco')
        ],
        value='without_ai_reco',
        description='Method:',
        layout=Layout(width='60%'),
        style={'description_width': '80px'}
    )

    # Create confirm button
    confirm_btn = widgets.Button(
        description='Confirm & Evaluate',
        button_style='success',
        layout=Layout(width='200px')
    )

    output = widgets.Output()

    def on_confirm(btn):
        with output:
            clear_output()
            method_name = method_dropdown.value

            if method_name == 'without_ai_reco':
                print("✅ Marked as: WITHOUT AI Recommendations")
            else:
                print("✅ Marked as: WITH AI Recommendations")

            # Evaluate and save
            evaluation_results = evaluate_and_save(
                final_ad=final_ad,
                background=background,
                product_image=config['product_image'],
                config=config,
                method_name=method_name
            )

            # Display scores prominently
            print("\n🎯 YOUR SCORES:")
            print(f"   Color Harmony:  {evaluation_results['color_harmony_score']}/100")
            print(f"   Aesthetic:      {evaluation_results['aesthetic_score']}/10")

            # Cleanup aesthetic model to free VRAM
            cleanup_aesthetic_model()

            print("\n✅ Evaluation complete! Generate more ads or run analyze_results()")

    confirm_btn.on_click(on_confirm)

    display(widgets.VBox([
        method_dropdown,
        confirm_btn,
        output
    ]))

    #=== END OF EVALUATION SECTION ====#



# Connect generate button
ui_controller.generate_btn.on_click(lambda btn: generate_advertisement())

print("\n✅ Generation system ready!")
print("📝 Configure all settings and click 'Generate Advertisement' when ready!")
print("\n💡 PRO TIPS:")
print("   • Lower guidance scale (3-5) for more creative, abstract results")
print("   • Higher guidance scale (10-15) for stricter prompt adherence")
print("   • Use seed values to reproduce exact results")
print("   • Generate multiple images to compare variations")
print("   • Experiment with different schedulers for unique aesthetics")
````

````python
# ============================================================================
# CELL 10: EVALUATION METRICS (LEVEL 1 ONLY)
# ============================================================================

import colorsys
import numpy as np
from sklearn.cluster import KMeans
from transformers import CLIPProcessor, CLIPModel
import torch
import json
from datetime import datetime

print("⏳ Loading evaluation libraries...")

# ============================================================================
# COLOR HARMONY CALCULATOR
# ============================================================================

def extract_dominant_colors_eval(image, n_colors=3):
    """Extract n dominant colors from image for evaluation"""
    try:
        img_array = np.array(image.resize((100, 100)))

        if len(img_array.shape) == 2:  # Grayscale
            img_array = np.stack([img_array]*3, axis=-1)

        if img_array.shape[2] == 4:  # RGBA
            mask = img_array[:, :, 3] > 128
            if mask.sum() == 0:
                return [(128, 128, 128)] * n_colors
            pixels = img_array[mask][:, :3]
        else:
            pixels = img_array.reshape(-1, 3)

        if len(pixels) < n_colors:
            return [(128, 128, 128)] * n_colors

        kmeans = KMeans(n_clusters=n_colors, random_state=42, n_init=10)
        kmeans.fit(pixels)

        return [tuple(map(int, c)) for c in kmeans.cluster_centers_]
    except Exception as e:
        print(f"   ⚠️ Color extraction error: {e}")
        return [(128, 128, 128)] * n_colors

def calculate_color_harmony_score(product_image, background_image):
    """
    Calculate color harmony between product and background (0-100)
    Based on color theory principles
    """
    print("   🎨 Extracting colors...")

    # Extract colors
    prod_colors = extract_dominant_colors_eval(product_image, n_colors=3)
    bg_colors = extract_dominant_colors_eval(background_image, n_colors=3)

    # Convert to HSV
    prod_hsv = [colorsys.rgb_to_hsv(r/255, g/255, b/255) for r,g,b in prod_colors]
    bg_hsv = [colorsys.rgb_to_hsv(r/255, g/255, b/255) for r,g,b in bg_colors]

    # Calculate harmony scores for all color pairs
    scores = []
    for ph, ps, pv in prod_hsv:
        for bh, bs, bv in bg_hsv:
            # Hue difference (0-180 degrees)
            hue_diff = abs(ph - bh) * 360
            if hue_diff > 180:
                hue_diff = 360 - hue_diff

            # Score based on color theory
            if hue_diff < 15:  # Monochromatic
                score = 95
            elif 25 < hue_diff < 35:  # Analogous
                score = 90
            elif 115 < hue_diff < 125:  # Triadic
                score = 85
            elif 175 < hue_diff < 185:  # Complementary
                score = 100
            elif 145 < hue_diff < 155:  # Split complementary
                score = 88
            else:
                # Gradual decrease from ideal harmony
                score = max(50, 80 - abs(hue_diff - 90)/2)

            scores.append(score)

    # Return average of top 3 harmony relationships
    final_score = np.mean(sorted(scores, reverse=True)[:3])
    return round(final_score, 2)

# ============================================================================
# AESTHETIC SCORE CALCULATOR
# ============================================================================

# Global variables for aesthetic model (load once, reuse)
aesthetic_model = None
aesthetic_processor = None

def load_aesthetic_model():
    """Load CLIP model for aesthetic scoring (only once)"""
    global aesthetic_model, aesthetic_processor

    if aesthetic_model is None:
        print("   ⏳ Loading aesthetic predictor (CLIP)...")
        aesthetic_model = CLIPModel.from_pretrained(
            "openai/clip-vit-base-patch32"
        ).to("cuda")
        aesthetic_processor = CLIPProcessor.from_pretrained(
            "openai/clip-vit-base-patch32"
        )
        print("   ✅ Aesthetic model loaded!")

def calculate_aesthetic_score(image):
    """
    Calculate aesthetic quality score (0-10)
    Uses CLIP to compare against aesthetic prompts
    """
    load_aesthetic_model()

    # Define aesthetic quality prompts
    good_prompts = [
        "a beautiful professional advertisement",
        "high quality commercial photography",
        "aesthetically pleasing design",
        "award winning advertisement"
    ]

    bad_prompts = [
        "ugly amateur advertisement",
        "low quality design",
        "poorly made ad"
    ]

    # Process image and prompts
    inputs = aesthetic_processor(
        text=good_prompts + bad_prompts,
        images=image,
        return_tensors="pt",
        padding=True
    ).to("cuda")

    # Calculate similarity scores
    with torch.no_grad():
        outputs = aesthetic_model(**inputs)
        probs = outputs.logits_per_image.softmax(dim=1)[0]

    # Score based on good prompts (0-10 scale)
    good_score = probs[:len(good_prompts)].sum().item()
    score = good_score * 10

    return round(score, 2)

def cleanup_aesthetic_model():
    """Free VRAM after evaluation"""
    global aesthetic_model, aesthetic_processor
    if aesthetic_model:
        del aesthetic_model
        del aesthetic_processor
        torch.cuda.empty_cache()
        aesthetic_model = None
        aesthetic_processor = None
        print("   🧹 Aesthetic model cleaned up")

# ============================================================================
# EVALUATION WRAPPER
# ============================================================================

def evaluate_and_save(final_ad, background, product_image, config, method_name):
    """
    Evaluate advertisement and save results

    Args:
        final_ad: Final composed advertisement (PIL Image)
        background: Generated background image (PIL Image)
        product_image: Product with bg removed (PIL Image)
        config: User configuration dict
        method_name: str - "without_ai_reco" or "with_ai_reco"

    Returns:
        dict with evaluation results
    """
    print("\n" + "="*70)
    print("📊 EVALUATING ADVERTISEMENT")
    print("="*70)

    # Color Harmony Score
    print("   🎨 Calculating color harmony...")
    try:
        harmony_score = calculate_color_harmony_score(product_image, background)
        print(f"   ✅ Color Harmony: {harmony_score}/100")
    except Exception as e:
        print(f"   ❌ Color harmony calculation failed: {e}")
        harmony_score = 0.0

    # Aesthetic Score
    print("   ✨ Calculating aesthetic score...")
    try:
        aesthetic_score = calculate_aesthetic_score(final_ad)
        print(f"   ✅ Aesthetic Score: {aesthetic_score}/10")
    except Exception as e:
        print(f"   ❌ Aesthetic calculation failed: {e}")
        aesthetic_score = 0.0

    # Compile results
    results = {
        'timestamp': datetime.now().isoformat(),
        'method': method_name,
        'product_description': config['product_description'][:100],
        'category': config['category'],
        'color_theme': config['color'],
        'design_motif': config['motif'],
        'color_harmony_score': harmony_score,
        'aesthetic_score': aesthetic_score,
        'guidance_scale': config.get('guidance_scale', 7.5),
        'seed': config.get('seed', -1)
    }

    # Save to JSON file
    timestamp_str = datetime.now().strftime('%Y%m%d_%H%M%S')
    filename = f"evaluation_{method_name}_{timestamp_str}.json"

    with open(filename, 'w') as f:
        json.dump(results, f, indent=2)

    print(f"\n💾 Evaluation saved to: {filename}")
    print("="*70 + "\n")

    return results

print("✅ Evaluation functions loaded!")



# ============================================================================
# CELL 11: ANALYSIS & VISUALIZATION
# ============================================================================

def analyze_results():
    """Analyze all evaluation results and generate comparison report"""
    import glob
    import pandas as pd
    import matplotlib.pyplot as plt
    from scipy.stats import ttest_ind

    print("\n" + "="*80)
    print("📊 ANALYZING AI RECOMMENDATIONS IMPACT")
    print("="*80)

    # Load all evaluation JSON files
    files = glob.glob("evaluation_*.json")

    if len(files) == 0:
        print("❌ No evaluation files found!")
        print("   Generate some ads first and evaluate them.")
        return

    results = []
    for f in files:
        with open(f, 'r') as file:
            results.append(json.load(file))

    df = pd.DataFrame(results)

    print(f"\n✅ Loaded {len(df)} evaluations")
    print(f"   - WITHOUT AI Reco: {len(df[df['method'] == 'without_ai_reco'])}")
    print(f"   - WITH AI Reco:    {len(df[df['method'] == 'with_ai_reco'])}")

    # Filter data
    without = df[df['method'] == 'without_ai_reco']
    with_ai = df[df['method'] == 'with_ai_reco']

    if len(without) == 0 or len(with_ai) == 0:
        print("\n⚠️ Need samples from BOTH methods for comparison!")
        print(f"   Current: {len(without)} without, {len(with_ai)} with AI")
        return

    # ========================================================================
    # STATISTICAL ANALYSIS
    # ========================================================================

    print("\n" + "="*80)
    print("📊 RESULTS COMPARISON")
    print("="*80)

    # Color Harmony
    print("\n🎨 COLOR HARMONY SCORES:")
    print("-" * 60)
    mean_without_h = without['color_harmony_score'].mean()
    std_without_h = without['color_harmony_score'].std()
    mean_with_h = with_ai['color_harmony_score'].mean()
    std_with_h = with_ai['color_harmony_score'].std()

    print(f"WITHOUT AI Reco:  {mean_without_h:.2f} ± {std_without_h:.2f}")
    print(f"WITH AI Reco:     {mean_with_h:.2f} ± {std_with_h:.2f}")

    diff_h = mean_with_h - mean_without_h
    pct_h = (diff_h / mean_without_h) * 100
    print(f"\n📈 IMPROVEMENT:    {diff_h:+.2f} points ({pct_h:+.1f}%)")

    if len(without) >= 2 and len(with_ai) >= 2:
        t_h, p_h = ttest_ind(with_ai['color_harmony_score'], without['color_harmony_score'])
        sig_h = "✅ SIGNIFICANT (p < 0.05)" if p_h < 0.05 else "❌ Not significant (p ≥ 0.05)"
        print(f"📊 Statistical Test: p = {p_h:.4f} - {sig_h}")

    # Aesthetic Score
    print("\n✨ AESTHETIC SCORES:")
    print("-" * 60)
    mean_without_a = without['aesthetic_score'].mean()
    std_without_a = without['aesthetic_score'].std()
    mean_with_a = with_ai['aesthetic_score'].mean()
    std_with_a = with_ai['aesthetic_score'].std()

    print(f"WITHOUT AI Reco:  {mean_without_a:.2f} ± {std_without_a:.2f}")
    print(f"WITH AI Reco:     {mean_with_a:.2f} ± {std_with_a:.2f}")

    diff_a = mean_with_a - mean_without_a
    pct_a = (diff_a / mean_without_a) * 100
    print(f"\n📈 IMPROVEMENT:    {diff_a:+.2f} points ({pct_a:+.1f}%)")

    if len(without) >= 2 and len(with_ai) >= 2:
        t_a, p_a = ttest_ind(with_ai['aesthetic_score'], without['aesthetic_score'])
        sig_a = "✅ SIGNIFICANT (p < 0.05)" if p_a < 0.05 else "❌ Not significant (p ≥ 0.05)"
        print(f"📊 Statistical Test: p = {p_a:.4f} - {sig_a}")

    # ========================================================================
    # VISUALIZATION
    # ========================================================================

    print("\n📊 Generating visualization...")

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    # Color Harmony Chart
    ax1 = axes[0]
    bars1 = ax1.bar(
        ['Without\nAI Reco', 'With\nAI Reco'],
        [mean_without_h, mean_with_h],
        yerr=[std_without_h, std_with_h],
        capsize=10,
        color=['#FF6B6B', '#4ECDC4'],
        alpha=0.8,
        edgecolor='black',
        linewidth=2
    )
    ax1.set_ylabel('Color Harmony Score', fontsize=13, fontweight='bold')
    ax1.set_ylim(0, 100)
    ax1.set_title('Color Harmony Score Comparison', fontsize=14, fontweight='bold')
    ax1.grid(axis='y', alpha=0.3, linestyle='--')

    # Add value labels on bars
    for bar in bars1:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.1f}',
                ha='center', va='bottom', fontweight='bold')

    # Aesthetic Score Chart
    ax2 = axes[1]
    bars2 = ax2.bar(
        ['Without\nAI Reco', 'With\nAI Reco'],
        [mean_without_a, mean_with_a],
        yerr=[std_without_a, std_with_a],
        capsize=10,
        color=['#FF6B6B', '#4ECDC4'],
        alpha=0.8,
        edgecolor='black',
        linewidth=2
    )
    ax2.set_ylabel('Aesthetic Score', fontsize=13, fontweight='bold')
    ax2.set_ylim(0, 10)
    ax2.set_title('Aesthetic Score Comparison', fontsize=14, fontweight='bold')
    ax2.grid(axis='y', alpha=0.3, linestyle='--')

    # Add value labels on bars
    for bar in bars2:
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.1f}',
                ha='center', va='bottom', fontweight='bold')

    plt.tight_layout()
    plt.savefig('ai_recommendations_impact.png', dpi=300, bbox_inches='tight')
    print("✅ Chart saved as 'ai_recommendations_impact.png'")
    plt.show()

    # ========================================================================
    # SUMMARY
    # ========================================================================

    print("\n" + "="*80)
    print("📝 SUMMARY")
    print("="*80)
    print(f"Sample Size:        {len(without)} products per method")
    print(f"Color Harmony:      {diff_h:+.2f} points improvement ({pct_h:+.1f}%)")
    print(f"Aesthetic Quality:  {diff_a:+.2f} points improvement ({pct_a:+.1f}%)")
    print("="*80)

print("✅ Analysis function ready!")
print("\n💡 USAGE:")
print("   1. Generate 10 ads WITHOUT AI reco (enter '1' when prompted)")
print("   2. Generate 10 ads WITH AI reco (enter '2' when prompted)")
print("   3. Run: analyze_results()")
````
